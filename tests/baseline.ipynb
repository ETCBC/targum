## The corpus

The corpus holds several Palestinian Targum manuscripts: Targum Pseudo-Jonathan, the Fragment
Targums, Cairo Genizah fragments, and Targum Neofiti. Neofiti is represented in two views, a
'base' text that keeps only unmarked segments, excluding variants and interlinear or in-line
additions, and a 'full' text that places every segment side by side. The text is Jewish
Palestinian Aramaic, carried in ETCBC transliteration and in Aramaic square script.

The object hierarchy is `text > book > chapter > verse > segment > word_group > word`, and `word`
is the slot type. A `lexeme` tier sits above `word` and is the one node type not contained within
the text tree, since the occurrences of a lexeme are distributed across the whole corpus.
`word_group` is included as a step toward full consistency across the hierarchy while the
underlying data model continues to develop. The segment `type` feature records the editorial
state of a run of text, with values including `unmarked_text`, `variant`, `addition`,
`editorial_deletion`, `broken_text`, and `interlinear`. Segments and words both carry `start_idx`
and `end_idx`, one-based character offsets into the `line` feature of the verse that contains
them. A segment's offsets are the minimum and the maximum of the offsets of the words it holds, so
the two tiers agree by construction.

Two features record how a word relates to the published line. `source` is `word` where
morphological analysis was placed directly on the word, `word_group` where the group carries a
surface form that the analysis does not individually cover, `none` where the manuscript records a
word that carries no morphological analysis at all, and `segment` where the word stands in for a
segment that no analysed word reaches, carrying that segment's own text so the segment can still
be published. `part` counts the parts of a word that an editorial boundary cuts in two. The two
parts share their `word_group` and its `id`, each sits in the segment that its own characters
fall in, and a word with `part` 1 and no sibling is whole. Because a split word group occupies
two nodes, the tests below that speak of a word group treat all nodes sharing one `id` as a
single unit.

## Baseline tests

Tests 1 to 5 move up through the object hierarchy and ask whether a given tier of nodes
reconstructs the tier above it, for example whether the segments of a verse reconstruct the verse
line. Test 4 also resolves the segment offsets against the line they index. Test 6 checks whether
the stored ordinals agree with the order in which the dataset publishes its nodes.

Each test reports its figures per corpus, because the five corpora differ substantially in how
much morphological analysis their source material provides.

### Reading the results

Tests 1, 3, 4, and 5 report figures that vary by corpus, reflecting properties of the source
material. Most of the failures fall into two categories, which the second table of test 1
isolates: surface forms whose morphological analysis covers only part of the word, and a
diacritic written one way at the word tier and another way at the word group tier.

Test 2 reports a figure that describes the publication boundary itself. A word that straddles two
editorial states is contained by only one segment, leaving the other segment without a word group
that spells it. The same effect limits the proportion of segment spans whose slice returns the
segment, and the proportion of verses whose segments spell the line, both reported in test 4.

## Interoperability tests

Tests 7 to 14 ask a different kind of question. Tests 1 to 6 measure a corpus whose figures
legitimately vary by manuscript, so they assert almost nothing and their table is their report.
Tests 7 to 14 state invariants, and each one is either true of the published dataset or is a
defect in it. They were written in response to an external review of the dataset, and most of
them failed on the build that the review read.

Test 7 asks whether the published transliteration is ETCBC. Lowercase is a private extension of
this project, in which a lowercase letter is the final form of its consonant. The BHSA writes
every consonant uppercase, because in Hebrew the final form is predictable from position. Here it
is not, which is why a final letter can appear mid-string, but it is recorded a second time in
the square-script features, so the ETCBC features are uppercased at the publication boundary.
Two integrity populations ride along with that check. The internal space that a multi-word
lexical entry once left in the consonantal form is now absent, so that figure is an invariant at
zero rather than a pinned ceiling. The Genizah lacuna mark is deliberate, which is a correction
to the first version of this notebook. All nine of its occurrences in the vocalised form are
words whose entire surface is a run of dots, standing in for a word the manuscript has lost, so
the invariant is that the mark never shares a word with a letter rather than that it never
occurs.

Test 8 asks whether `lex` is a lexeme. It counts the distinct lexemes of each corpus and how many
of them appear in neither Neofiti corpus, those being the two whose lexeme the view layer drew
from the lemma tree rather than from the surface column of the word, and it reads the same fact a
second way as the words whose `lex` equals their `cons`. Its inventory figures are pinned,
because they were measured against the regenerated view bodies before the rebuild and are
therefore a prediction that the rebuild has to reproduce. Its node counts are reported, because a
node count is not a set size.

Test 9 covers two features that carried something other than what they are named for: `gloss`,
whose fallback handed it the lexical category published beside it as `description`, and the word
half of `text`, which was a byte-for-byte copy of `cons_utf8` published under the name of the
section feature of the corpus.

Test 10 is the section census, and the figures it pins are a documented ceiling rather than a
target. A chapter 0 is the corpus stating that an unlocalised fragment names its book and not its
chapter, and turning it into a chapter 1 would be a fabrication. It passes, and is the regression
guard.

Test 11 counts every node feature by the node type its occurrences fall on, and asserts that the
only names present at two tiers are those that measure one quantity at two tiers. It is the
generic form of what test 9 found by hand, and it fails on the next such feature without anyone
having to anticipate it.

Test 12 covers the features that this version publishes and the previous one did not, together
with the larger set of candidates that were considered and withheld. For each feature present it
asserts the property that makes the feature worth publishing rather than merely its presence, and
for each feature withheld it asserts absence. Both directions matter. A feature read from the
wrong tier returns `None` without raising, which is the sharpest reason to assert the tier and
not the name, and a withheld feature that returned in a later rebuild would be a silent change to
a citable dataset. The strongest of the assertions are structural. For every word that has a
lexeme, `lex` must begin with `lex0` and the remainder must be drawn from `[`, `/` and `=` and
nothing else, which catches a marker on the wrong base, a base that has drifted, and a stray
character in one statement. For the lexeme tier, `rank_lex` must be a total ranking covering 1 to
the number of lexemes without a gap, because that property is what allows the tier to be keyed on
it and is why no separate lexeme identifier is published.

Test 13 asks where the lexeme and the surface disagree, and decomposes the words that share no
letter with their own lexeme into the causes that account for them. Two causes that earlier
versions could name are no longer separable from published features, so those words now fall
through to the remaining rows and the unexplained row is reported rather than pinned.

Test 14 is the parallel witness edge, `crossref`, which is a feature of the core data rather than
a module to be fetched alongside it. Two verses are parallel when they carry the same book,
chapter and verse, so the relation is given by the data and is not inferred, and there is no
similarity threshold and no choice of method. The test asserts that the edge is that relation: it
is symmetric in both directions, never joins two addresses, never leaves two nodes at one address
unjoined, and is labelled correctly across all three of `witness`, where the manuscripts differ
and the relation is a parallel, `view`, where one manuscript is read two ways, which is Neofiti 1
and nothing else, and `same_witness`, where one manuscript under one view holds two verse records
at one address, which three Genizah fragments do. It ends in four worked examples, because the
purpose of the edge is the questions it makes inexpensive.

Each of these reports through `check`, which records a failure and continues, so that one run
produces every figure. The last cell is the gate.

Three of the review's findings proved to be wrong about their own cause, and the cells say so
where they were measured. The first is the lacuna mark described above. The second is the isolect
column, which keys the hierarchy of isolects rather than that of dialects and was therefore the
hierarchy of the source rather than an off-by-one error. The third is the lemma fallback on
`cons`, which was an unsound expression that never once supplied a value in the published corpus.
What check 7c of the SQL baseline is really measuring is the whitespace model, and 2,757 of its
2,835 failures are the deferred decision about the slot model for which test 7 pins the starting
point.


```python
# Loads the dataset and reports its provenance. The version string, the version of the XML the
# dataset was compiled from, and the compilation timestamp are read from the metadata of the
# `otype` feature, which every Text-Fabric dataset carries. The node counts printed here are the
# quantities every later test is measured against.
#
# The corpus is loaded through context-fabric rather than through the advanced application layer
# of Text-Fabric, and three consequences follow from that.
#
# First, the corpus is addressed by path rather than by organisation and repository.
# `cfabric.Fabric(locations, modules)` reads the versioned dataset directory directly, so the
# notebook depends neither on an environment resolving a repository name nor on an environment
# file existing at all. That is what allows this file to run unchanged in the public repository.
#
# Second, context-fabric loads from its own binary cache, which retains feature values and
# discards the header block that carries the metadata. The metadata available through the library
# is therefore empty, and the provenance reported below depends on it, so `tf_meta` reads the
# header out of the shipped feature file itself. That is not a workaround. It reads the artefact
# being published rather than a library's parse of it, so it cannot disagree with what ships. The
# binary cache also holds no staleness check against the feature files beside it, which is why
# the pipeline deletes it after every compile, and reading the files directly is immune to the
# question either way.
#
# Third, there is no advanced application layer, so the helper defined below replaces it and
# prints rather than rendering markup, which keeps the output legible in a terminal and in a
# difference listing.
#
# The version is deliberately not pinned in this notebook. The declared version of the dataset is
# read from the application configuration, which is the same value the compiler writes into the
# metadata of every feature, so this notebook cannot describe an older build than the repository
# ships. An earlier version pinned the string as a literal and continued to report itself as
# current across a rebuild, which would have described a superseded corpus throughout.

import os
from pathlib import Path

import yaml
from dotenv import load_dotenv

import cfabric

REPO_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "config/xml.yml").exists())
load_dotenv(REPO_ROOT / ".env")

TF_VERSION = yaml.safe_load((REPO_ROOT / "app/config.yaml").read_text())["provenanceSpec"]["version"]
TF_DIR = REPO_ROOT / "tf" / TF_VERSION

# Kept for the provenance line only. Loading no longer goes through them.
ORG = os.environ.get("TF_ORG") or "etcbc"
REPO = os.environ.get("TF_REPO") or "targum"

# The silence level is passed to both calls, because the loader takes its own. Without it the
# library logs one line per feature as it loads, which is longer than most of the output below
# and conveys nothing that the feature inventory does not. The accepted levels are "verbose",
# "auto", "terse" and "deep".
CF = cfabric.Fabric(locations=str(REPO_ROOT / "tf"), modules=TF_VERSION, silent="deep")
api = CF.loadAll(silent="deep")
F, E, L, T, S = api.F, api.E, api.L, api.T, api.S
Fs, Es = api.Fs, api.Es
TF_FEATURES = set(CF.features)


def tf_meta(feature="otype"):
    """The `@`-metadata of one feature, read from the shipped `.tf` file.

    cfabric serves feature data out of its `.cfm` cache, which stores values and drops the `@`
    header, so `CF.features[f].metaData` is empty. Every `.tf` file opens with its own metadata
    block, terminated by the first line that does not start with `@`, so reading it is four lines
    and is authoritative about the artefact rather than about a cache of it.
    """
    out = {}
    path = TF_DIR / f"{feature}.tf"
    if not path.exists():
        return out
    with path.open(encoding="utf8") as fh:
        for line in fh:
            if not line.startswith("@"):
                break
            if "=" in line:
                key, value = line[1:].rstrip("\n").split("=", 1)
                out[key] = value
    return out


def plain(node, withNodes=True):
    """Print a node and the words it contains, in place of the app layer's `pretty`.

    One line of heading and one line per word: the vocalised surface, the lexeme, the part of
    speech and `source`, which together are what the failing cells below are inspected for.
    """
    otype = F.otype.v(node)
    head = f"{otype} {node}" if withNodes else otype
    bits = []
    for name in ("id", "reading", "type", "verse", "chapter", "book", "text"):
        if name in TF_FEATURES:
            value = Fs(name).v(node)
            if value is not None:
                bits.append(f"{name}={value!r}")
    print(f"{head}  {'  '.join(bits)}")
    for w in L.d(node, otype="word") if otype != "word" else (node,):
        print(f"    {w:>8}  {F.voc.v(w)!r:<22} lex={F.lex.v(w)!r:<14} "
              f"sp={F.sp.v(w)!r:<8} source={F.source.v(w)!r}")


meta = tf_meta("otype")
```

```python
# Checks the dataset's provenance against the files it was compiled from. The dataset is stale
# unless both checks below hold.
#
#   1  The xmlVersion metadata against the newest directory under `xml/`. A dataset compiled from
#      an older XML directory than the one currently present is behind by at least that much.
#   2  The feature descriptions against `config/xml.yml`. A mismatch here means the dataset was
#      compiled before the current feature documentation was written.
#
# A failure in either check means the figures below describe an outdated build of the dataset.

import re
import yaml

FRESH = True


def stale(what, got, want):
    global FRESH
    FRESH = False
    print(f"  STALE  {what}\n           dataset: {got}\n           source:  {want}")


xml_versions = sorted(p.name for p in (REPO_ROOT / "xml").iterdir() if p.is_dir())
newest_xml = xml_versions[-1] if xml_versions else None
print(f"xmlVersion {meta.get('xmlVersion')}  written {meta.get('dateWritten')}")
print(f"newest xml/ {newest_xml}  (of {len(xml_versions)})")
if newest_xml and meta.get("xmlVersion") != newest_xml:
    stale("the dataset was compiled from an older XML than xml/ holds",
          meta.get("xmlVersion"), newest_xml)

declared = yaml.safe_load((REPO_ROOT / "config/xml.yml").read_text())["featureDescriptions"]


def norm(s):
    return re.sub(r"\s+", " ", (s or "")).strip()


for name, want in declared.items():
    if name not in TF_FEATURES:
        continue
    # `tf_meta` reads the header of the feature file directly. The metadata exposed by the
    # library is empty, because it serves from a binary cache that retains values and discards
    # the header block, so this
    # check would have reported all 58 descriptions as stale.
    got = tf_meta(name).get("description")
    if norm(got) != norm(want):
        stale(f"feature {name!r} carries a description config/xml.yml no longer declares",
              norm(got)[:110], norm(want)[:110])

counts = {t: n for t, n in ((t, len(list(F.otype.s(t)))) for t in F.otype.all)}
print("nodes:", ", ".join(f"{t} {counts[t]:,}" for t in F.otype.all))

print()
print("dataset is current" if FRESH else "DATASET IS STALE, rebuild before reading on")
assert FRESH, "stale dataset: see the report above"
```

```python
# Reconciles the feature inventory the metadata PROMISES against the features the dataset
# actually ships, and defines the soft check the new tests below report through.
#
# The staleness cell above compares each `config/xml.yml` description against the dataset, but it
# skips a feature the dataset does not have (`if feat is None: continue`), so a description for a
# feature that was never emitted is never checked and never noticed. `g_lex` and `g_lex_utf8` have
# been declared since the file was written and have never been published. This cell closes that
# hole from the other side.
#
# `check` is the reporting convention for every test added below. The six original tests read as
# measurements of a corpus whose figures legitimately vary by manuscript, so they assert almost
# nothing. The tests added here state invariants instead, and an invariant that fails should not
# stop the notebook: the whole point of running it is to get every figure in one pass. So a
# failed check is recorded and printed, the run continues, and the last cell is the gate.

violations = []


def check(name, got, want, note=""):
    ok = got == want
    if not ok:
        violations.append((name, got, want, note))
    tail = "" if ok else f"  expected {want:,}"
    print(f"  {'ok  ' if ok else 'FAIL'}  {name:<46} {got:>9,}{tail}")
    if not ok and note:
        print(f"          {note}")
    return ok


# `TF.features` also holds the nine `__`-prefixed stores Text-Fabric computes for itself, and
# `otype`/`oslots` are the dataset's skeleton rather than annotations, so the shipped inventory is
# the node features plus the edge features, less those two.
emitted = (set(api.Fall()) | set(api.Eall())) - {"otype", "oslots"}

# `featureDescriptions` also documents the NODE TYPES, and a node type has no `.tf` file of
# its own, so it is not an undelivered feature. The set of node types is derived rather than
# listed. An earlier version held a literal pair of names, so the lexeme node type, correctly
# described in the metadata file and the one node type this version adds, read as a description
# that no
# feature answers. A node type that is ALSO a feature (`text`, `book`, `chapter`, `verse`,
# `segment`) stays in `promised`, because it does ship a file and the description is that
# feature's. Deriving it means the next node type does not need anyone to think of it.
NODE_TYPE_LABELS = set(F.otype.all) - emitted
promised = set(declared) - NODE_TYPE_LABELS
undelivered = sorted(promised - emitted)
undescribed = sorted(emitted - set(declared))

print(f"config/xml.yml describes {len(declared)} names; the dataset ships {len(emitted)} features")
print()
print(f"described but never emitted ({len(undelivered)}): {', '.join(undelivered) or 'none'}")
print(f"emitted but never described ({len(undescribed)}): {', '.join(undescribed) or 'none'}")
print()
check("features described but not emitted", len(undelivered), 0,
      "a description in config/xml.yml that no .tf file answers; the staleness cell "
      "cannot see these because it skips a feature the dataset lacks")
check("features emitted but not described", len(undescribed), 0,
      "a published feature carrying no documented meaning")
```

```python
# Reports the feature inventory and the distribution of the three features that the tests below
# depend on. `source` determines which words carry a morphological analysis, `reading` identifies
# the readings that compete for one position, and the segment `type` feature records the editorial
# state of a run of the line.

FEATURES = sorted(f for f in TF_FEATURES if not f.startswith("o") and f != "otext")
for otype in F.otype.all:
    sample = F.otype.s(otype)[:400]
    present = sorted(f for f in FEATURES
                     if hasattr(F, f) and any(Fs(f).v(n) is not None for n in sample))
    print(f"{otype:<11} {', '.join(present)}")
print()
print("source     ", dict(F.source.freqList()))
print("reading    ", dict(F.reading.freqList()))
print("segment    ", dict(F.type.freqList()))
print()
for fmt in sorted(T.formats):
    print(f"  {fmt:<16} {T.text(range(1, 9), fmt=fmt)}")
```

```python
# Defines the material shared by every test below.
#
# `letters` is the comparison class used throughout. It strips whitespace so that strings drawn
# from different tiers of the corpus, which do not always agree on how much space they record, can
# still be compared fairly.
#
# `verse_nodes` keys each verse by the pair (manuscript, uid). The `uid` feature is unique within
# one corpus and repeats across corpora, so the manuscript name is needed to make the key unique.
# Two verses of the Genizah material occupy more than one verse node, because the order in which
# the dataset publishes their words does not keep the rows of those verses contiguous. The tests
# below aggregate over the list of nodes that share a key.
#
# `table` prints one row per corpus with a total, and takes derived percentage columns as
# (label, numerator, denominator) triples.

import re
from collections import Counter, defaultdict

_WS = re.compile(r"\s+")


def letters(s):
    return _WS.sub("", s or "")


CORPORA = ["cairo_genizah", "frag_targ", "neofiti_base_raw", "neofiti_full", "psj"]
_BY_NAME = {"neofiti_base": "neofiti_base_raw", "neofiti_full": "neofiti_full",
            "pseudo_jonathan": "psj"}


def corpus_of(text_node):
    name = F.text.v(text_node)
    if name.startswith("cairo"):
        return "cairo_genizah"
    if name.startswith("fragment_targum"):
        return "frag_targ"
    return _BY_NAME[name]


# The key of this map was changed. It was keyed on the manuscript and the verse identifier,
# and that identifier is now withheld
# from the publication contract: a monad id collides almost completely across the three family
# tables, so a consumer joining on it silently read the wrong row. The verse tier therefore
# carries no identifier at all, and the feature is null on all 19,595 nodes, so every key
# collapsed to the manuscript alone,
# this dict held 46 buckets instead of 19,589, and every TF-side figure below was computed over
# 46 pseudo-verses while reporting itself as a corpus-wide measurement.
#
# The replacement is the published identity of a verse, which is its canonical address together
# with the manuscript. It is strictly better than the key it replaces. There are 19,589 addresses
# over 19,589 verses, one identifier each, against 19,587 for the manuscript and identifier pair,
# which merges the two verses whose rows carry two addresses. It is also the key on which the
# parallel witness edge is measured, where it collides no times.
#
# A verse still reaches TF as more than one verse node wherever the publication order does not
# keep its rows contiguous, which is why the value here is a LIST. Three verses do, in the Geniza
# fragments `dd` and `ff`: 19,595 nodes over 19,589 addresses. The interleaving is with a
# neighbouring verse rather than a second transcription, since every node at one address carries
# one
# `line` value, and Exodus 15:12 and 17:12 of `ff` interleave with each other.
VERSE_KEY_COLUMNS = ("text_text", "book_book", "chapter_chapter", "verse_verse")

verse_nodes = defaultdict(list)
corpus_of_verse = {}
for _t in F.otype.s("text"):
    _name, _corpus = F.text.v(_t), corpus_of(_t)
    for _v in L.d(_t, otype="verse"):
        _k = (_name, F.book.v(L.u(_v, otype="book")[0]),
              F.chapter.v(L.u(_v, otype="chapter")[0]), F.verse.v(_v))
        verse_nodes[_k].append(_v)
        corpus_of_verse[_k] = _corpus


def table(title, rows, cols, pct=(), key="corpus"):
    keys = [k for k in CORPORA if k in rows] + [k for k in rows if k not in CORPORA]
    totals = Counter()
    for k in keys:
        totals.update(rows[k])
    head = [key, *cols, *(lbl for lbl, _, _ in pct)]
    body = []
    for k in [*keys, "TOTAL"]:
        r = totals if k == "TOTAL" else rows[k]
        body.append([k, *(f"{r[c]:,}" for c in cols),
                     *(f"{100.0 * r[n] / r[d]:.2f}" if r[d] else "-" for _, n, d in pct)])
    w = [max(len(str(x)) for x in col) for col in zip(head, *body)]
    rule = "  ".join("-" * i for i in w)
    print(title)
    print("  " + "  ".join(h.rjust(i) for h, i in zip(head, w)))
    print("  " + rule)
    for line in body[:-1]:
        print("  " + "  ".join(str(x).rjust(i) for x, i in zip(line, w)))
    print("  " + rule)
    print("  " + "  ".join(str(x).rjust(i) for x, i in zip(body[-1], w)))


results = {}
print(f"{len(verse_nodes):,} verses distributed over "
      f"{sum(len(v) for v in verse_nodes.values()):,} verse nodes")
```

```python
# TEST 1. Composition of a word group from its words.
#
# Method: for every word group, concatenate the `voc` feature of the `word` nodes it contains, in
# slot order, and compare the result to the group's own `g_voc` feature under the letters
# comparison.
#
# A word group is treated here by its `id`, not as a single node. Where an editorial boundary cuts
# a word in two, each piece sits in a different segment, so the dataset publishes a second
# `word_group` node for it. Both nodes carry the whole source word as `g_voc`, and each holds only
# its own words. Reading such a group node by node would report it as failing twice for no real
# reason. Reading it by `id`, which is what this test does, states the intended claim: the
# morphological analysis of a word accounts for that word.
#
# What it measures: whether the morphological decomposition accounts for the surface form it was
# derived from. A group whose words spell it exactly is one where every character of the
# manuscript word is claimed by some morpheme.
#
# Words that the source never analysed at all, where `source` is `none` and which number 1,304,
# have no word group to
# spell: the dataset publishes each under a group node whose `id` and `g_voc` are both empty.
# Keying by `id` therefore collected every such word of a verse into ONE bucket whose target was
# the empty string, and that bucket failed once per verse that held any. Those words are excluded
# here, and a bucket the exclusion empties is dropped. It removes 602 phantom buckets carrying
# 599 of the 723 failures this test used to report: `groups` 361,013 -> 360,411, `differ`
# 723 -> 124, `words` 495,216 -> 493,912, and `%` 99.80 -> 99.97. `analysed`, `analysed_spelled`
# and `% of analysed` do not move at all (354,730 / 354,608 / 99.97), because a groupless bucket
# was never in the answerable population, which is why the proportion of analysed words is the
# honest column
# all along, and why the `%` column now agrees with it.
#
# Interpretation: `analysed` restricts the population to the groups whose words all carry
# morphological analysis, the set the decomposition can be evaluated against. The second table
# classifies the failures. `reordered` counts groups where the same letters appear in a different
# order, an alignment fault. `words_shorter` counts groups where the analysis covers only part of
# the surface form, a property of the source material. `dot_vs_mark` counts groups where the two
# strings differ only in whether a Genizah diacritic is written as a full stop or as the combining
# mark U+0740.

test1 = defaultdict(Counter)
group_failures = []
for _k, _vns in verse_nodes.items():
    _c = corpus_of_verse[_k]
    # The pieces of one word group, in node order, keyed by the group's own id.
    _by_id = defaultdict(list)
    for _v in _vns:
        for _g in L.d(_v, otype="word_group"):
            _by_id[F.id.v(_g)].append(_g)
    for _gid, _nodes in _by_id.items():
        ws = [w for _g in _nodes for w in L.d(_g, otype="word")
              if F.source.v(w) != "none"]
        if not ws:
            continue
        built = "".join(F.voc.v(w) or "" for w in ws)
        target = F.g_voc.v(_nodes[0]) or ""
        analysed = {F.source.v(w) for w in ws} == {"word"}
        ok = letters(built) == letters(target)
        _r = test1[_c]
        _r["groups"] += 1
        _r["words"] += len(ws)
        _r["split_groups"] += len(_nodes) > 1
        _r["words_spell_group"] += ok
        _r["differ"] += not ok
        _r["analysed"] += analysed
        _r["analysed_spelled"] += analysed and ok
        if not ok:
            group_failures.append((_c, _nodes[0], target, built))
            _b, _t = letters(built), letters(target)
            _r["reordered"] += sorted(_b) == sorted(_t)
            _r["words_shorter"] += len(_b) < len(_t)
            _r["words_longer"] += len(_b) > len(_t)
            _r["dot_vs_mark"] += _b.replace("݀", ".") == _t

table("TEST 1  the words of a word group spell the group", test1,
      ["groups", "split_groups", "words", "words_spell_group", "differ", "analysed",
       "analysed_spelled"],
      pct=[("%", "words_spell_group", "groups"), ("% of analysed", "analysed_spelled", "analysed")])
print()
table("TEST 1  the composition of the failures", test1,
      ["differ", "reordered", "words_shorter", "words_longer", "dot_vs_mark"])
results["1 words compose the word group"] = (sum(test1[c]["groups"] for c in CORPORA),
                                             sum(test1[c]["words_spell_group"] for c in CORPORA))
```

```python
# Displays four of the word groups that test 1 reports as failing, followed by one of them
# rendered node by node by `plain`. The printed line gives the published surface form, the form the
# words spell, and the analysis of each word, so the disagreement can be inspected directly. The
# second table of test 1 already summarises the failures as a population.

for _c, _g, _target, _built in group_failures[:4]:
    print(f"{_c:<16} word group {_g}: published {_target!r}, its words spell {_built!r}")
    print(f"{'':<16}   words: "
          + str([(F.voc.v(w), F.lex.v(w), F.sp.v(w), F.source.v(w)) for w in L.d(_g, otype='word')]))

_c, _g, _target, _built = next((f for f in group_failures if f[0] == "neofiti_full"),
                               group_failures[0])
print(f"\n{_c}: word group {_g}")
plain(_g, withNodes=True)
```

```python
# TEST 2. Composition of a segment from the words it contains.
#
# Method: for every `segment` node, concatenate the `voc` feature of the `word` nodes it contains
# and compare the result to the segment's own `segment` feature under the letters comparison.
#
# The word tier is the correct level to measure at here. `g_voc` records the whole source word, so
# a word cut by an editorial boundary contributes only its own characters to each side, and a
# segment holding half a word would be compared against the whole word and fail by construction.
# The word tier carries exactly the characters of the segment it sits in. `whole_groups` keeps the
# group-level comparison for reference, and it fails wherever a word straddles two editorial
# states.
#
# `all_readings` uses every word in the segment. `base_reading` uses only the words of groups
# whose `reading` is 0, the underlying reading the first scribe wrote. A segment holding no
# variant satisfies both identically.
#
# What it measures: whether the word tier accounts for the editorial segment that contains it.
# Each word is attributed to the segment its own span falls in, and a word whose span crosses a
# segment boundary is split so that each piece belongs to its own segment.

test2 = defaultdict(Counter)
for _k, _vns in verse_nodes.items():
    _c = corpus_of_verse[_k]
    for _v in _vns:
        for _s in L.d(_v, otype="segment"):
            gs = L.d(_s, otype="word_group")
            ws = L.d(_s, otype="word")
            target = letters(F.segment.v(_s))
            every = letters("".join(F.voc.v(w) or "" for w in ws))
            base = letters("".join(F.voc.v(w) or "" for g in gs if F.reading.v(g) == 0
                                   for w in L.d(g, otype="word")))
            whole = letters("".join(F.g_voc.v(g) or "" for g in gs))
            _r = test2[_c]
            _r["segments"] += 1
            _r["with_words"] += bool(ws)
            _r["no_letters"] += not target
            _r["has_variants"] += any(F.reading.v(g) != 0 for g in gs)
            _r["filler_only"] += bool(ws) and all(F.source.v(w) == "segment" for w in ws)
            _r["all_readings"] += every == target
            _r["base_reading"] += base == target
            _r["either"] += (every == target or base == target)
            _r["whole_groups"] += whole == target

table("TEST 2  the words of a segment spell the segment", test2,
      ["segments", "with_words", "no_letters", "has_variants", "filler_only",
       "all_readings", "base_reading", "either", "whole_groups"],
      pct=[("% either", "either", "segments"),
           ("% whole groups", "whole_groups", "segments")])
results["2 words compose the segment"] = (sum(test2[c]["segments"] for c in CORPORA),
                                          sum(test2[c]["either"] for c in CORPORA))
```

```python
# TEST 3. Reconstruction of the verse line from the word group tier.
#
# Method: concatenate the `g_voc` feature of every word group of a verse, in publication order,
# and compare the result to the verse's `line` feature under the letters comparison. Variant
# readings are included, because the published line carries them inline.
#
# Each word group is counted once, by its `id`. A group that an editorial boundary cuts in two
# publishes as two nodes carrying the same `g_voc`, so concatenating over nodes instead of ids
# would double-count that word's surface form and fail the verse for an unrelated reason.
#
# What it measures: whether the surface forms recorded at the word group tier account for the
# verse line the dataset publishes. This is the strongest statement available at that tier, and it
# is independent of the morphological decomposition tested in test 1.

test3 = defaultdict(Counter)
for _k, _vns in verse_nodes.items():
    _r = test3[corpus_of_verse[_k]]
    line = max({F.line.v(v) or "" for v in _vns})
    _seen, _built = set(), []
    for _v in _vns:
        for _g in L.d(_v, otype="word_group"):
            _gid = F.id.v(_g)
            if _gid in _seen:
                continue
            _seen.add(_gid)
            _built.append(F.g_voc.v(_g) or "")
    _r["verses"] += 1
    _r["word_groups"] += len(_seen)
    _r["groups_spell_line"] += letters("".join(_built)) == letters(line)

table("TEST 3  the word groups of a verse spell the verse line", test3,
      ["verses", "word_groups", "groups_spell_line"],
      pct=[("%", "groups_spell_line", "verses")])
results["3 word groups compose the verse line"] = (sum(test3[c]["verses"] for c in CORPORA),
                                                   sum(test3[c]["groups_spell_line"] for c in CORPORA))
```

```python
# TEST 4. Reconstruction of the verse line from the segment tier.
#
# Method: concatenate the `segment` feature of every `segment` node of a verse, in publication
# order, and compare the result to the verse's `line` feature under the letters comparison.
# Variant segments are included, because the published line carries the competing readings
# inline. Restricting the concatenation to segments that are not variants would still recover no
# verse that the full concatenation loses. `utf8_spell_line` repeats the comparison in Aramaic
# square script. `repeated` counts the segments a verse publishes more than once under one `id`,
# so the concatenation emits their text twice.
#
# The same loop resolves the segment offsets against the line, which the cell below reports. The
# line cut at a segment's span is compared to the segment's own string, and the spans of a verse
# are then examined as a partition. They must not overlap, the material between two consecutive
# spans must be whitespace only, and the material before the first span and after the last span
# must also be whitespace. A verse that satisfies all three conditions has segments that tile the
# line. Whether the spans also ascend in publication order is test 6's question, asked of every
# tier at once. The word tier carries its own `start_idx` and `end_idx`, resolved two cells below.
# A segment's span is the minimum and the maximum of the spans of the words it holds, so the two
# tiers agree by construction, and the word-level statement is the stronger of the two.
#
# What it measures: whether the editorial tier accounts for the verse line the dataset publishes.
# Test 3 asks that question of the word group tier and test 5 asks it of the word tier. The
# segment is the tier between them, and the only one of the three whose strings are runs of the
# line itself, independent of morphology.
#
# Interpretation: the second table restricts the population by the morphological analysis
# underneath. A segment carries morphological analysis when every word it holds has `source`
# `word`. A verse is analysed when every one of its segments does. That is the population the
# decomposition can be evaluated against, and it is reported beside the overall figure and beside
# the rate for the verses that hold an unanalysed segment, so the three can be read separately, as
# tests 1 and 5 do at their own tiers. The third table attributes the failures.
# `text_differs_from_its_span` counts the verses holding a segment whose stored string does not
# match the slice of the line named by its own offsets, the same containment boundary that test 2
# measures from the other side. `spans_still_cover_the_line` counts the failures whose geometry
# still returns the whole line, meaning the loss originates in the segment strings while the spans
# remain correct. The fourth table gives the proportion of the letters of each line that the
# segments emit at all. This proportion is high everywhere, so the shortfall consists of many
# small local losses spread across the text.

test4 = defaultdict(Counter)
segment_failures = []
for _k, _vns in verse_nodes.items():
    _r = test4[corpus_of_verse[_k]]
    line = max({F.line.v(v) or "" for v in _vns})
    line_utf8 = max({F.line_utf8.v(v) or "" for v in _vns})
    segs = [s for v in _vns for s in L.d(v, otype="segment")]
    target = letters(line)
    built = letters("".join(F.segment.v(s) or "" for s in segs))
    ok = built == target
    analysed, spans = [], set()
    for s in segs:
        ws = L.d(s, otype="word")
        analysed.append(bool(ws) and all(F.source.v(w) == "word" for w in ws))
        a, b = F.start_idx.v(s), F.end_idx.v(s)
        if a is None:
            _r["segment_no_offset"] += 1
            continue
        _r["segment_offset"] += 1
        _r["segment_slice_is_segment"] += line[a - 1:b] == (F.segment.v(s) or "")
        spans.add((a, b))
    every_analysed = bool(analysed) and all(analysed)
    _r["verses"] += 1
    _r["segments"] += len(segs)
    _r["repeated"] += len(segs) - len({F.id.v(s) for s in segs})
    _r["segments_spell_line"] += ok
    _r["utf8_spell_line"] += letters(
        "".join(F.segment_utf8.v(s) or "" for s in segs)) == letters(line_utf8)
    _r["line_letters"] += len(target)
    _r["segment_letters"] += len(built)
    _r["with_morphology"] += sum(analysed)
    _r["without_morphology"] += len(analysed) - sum(analysed)
    _r["analysed_verses"] += every_analysed
    _r["analysed_spell_line"] += every_analysed and ok
    _r["rest_verses"] += not every_analysed
    _r["rest_spell_line"] += (not every_analysed) and ok
    if spans:
        # A verse whose rows are not contiguous under the publication order is emitted as more
        # than one verse node, and a segment spanning the break is emitted once per node. The
        # partition is a statement about distinct spans, so they are collected as a set.
        ordered = sorted(spans)
        disjoint = all(nxt[0] > cur[1] for cur, nxt in zip(ordered, ordered[1:]))
        gaps_ws = all(not line[cur[1]:nxt[0] - 1].strip() for cur, nxt in zip(ordered, ordered[1:]))
        edges_ws = not line[:ordered[0][0] - 1].strip() and not line[ordered[-1][1]:].strip()
        _r["verses_with_spans"] += 1
        _r["spans_disjoint"] += disjoint
        _r["gaps_are_whitespace"] += gaps_ws
        _r["edges_are_whitespace"] += edges_ws
        _r["partitions_the_line"] += disjoint and gaps_ws and edges_ws
    if not ok:
        segment_failures.append((corpus_of_verse[_k], _k, target, built))
        _r["differ"] += 1
        _r["reordered"] += sorted(built) == sorted(target)
        _r["segments_shorter"] += len(built) < len(target)
        _r["segments_longer"] += len(built) > len(target)
        _r["text_differs_from_its_span"] += any(
            F.start_idx.v(s) is not None
            and line[F.start_idx.v(s) - 1:F.end_idx.v(s)] != (F.segment.v(s) or "")
            for s in segs)
        _r["spans_still_cover_the_line"] += letters(
            "".join(line[a - 1:b] for a, b in sorted(spans))) == target
        _r["segment_without_a_span"] += any(F.start_idx.v(s) is None for s in segs)

table("TEST 4  the segments of a verse spell the verse line", test4,
      ["verses", "segments", "repeated", "segments_spell_line", "utf8_spell_line"],
      pct=[("%", "segments_spell_line", "verses"), ("% utf8", "utf8_spell_line", "verses")])
print()
table("TEST 4  by the morphology under the segment", test4,
      ["segments", "with_morphology", "without_morphology", "analysed_verses",
       "analysed_spell_line", "rest_verses", "rest_spell_line"],
      pct=[("% segments analysed", "with_morphology", "segments"),
           ("% of analysed verses", "analysed_spell_line", "analysed_verses"),
           ("% of the rest", "rest_spell_line", "rest_verses")])
print()
table("TEST 4  the composition of the failures", test4,
      ["differ", "reordered", "segments_shorter", "segments_longer",
       "text_differs_from_its_span", "spans_still_cover_the_line", "segment_without_a_span"])
print()
table("TEST 4  how much of each line the segments emit", test4,
      ["line_letters", "segment_letters"],
      pct=[("% of the letters", "segment_letters", "line_letters")])
results["4 segments compose the verse line"] = (sum(test4[c]["verses"] for c in CORPORA),
                                                sum(test4[c]["segments_spell_line"]
                                                    for c in CORPORA))
```

```python
# The same segments, now read as geometry through the two tables the offsets support. `% with a
# span` gives the population that the second percentage is taken over. A segment whose slice
# returns its own text is one whose characters all belong to words that the segment holds, the
# same containment property that test 2 measures from the other side. A verse that partitions is
# one whose segments tile its line with nothing but whitespace between them.

table("TEST 4  the segment spans, resolved against the line", test4,
      ["segments", "segment_offset", "segment_no_offset", "segment_slice_is_segment"],
      pct=[("% with a span", "segment_offset", "segments"),
           ("% slice is the segment", "segment_slice_is_segment", "segment_offset")])
print()
table("TEST 4  the segment spans as a partition of the line", test4,
      ["verses", "verses_with_spans", "spans_disjoint", "gaps_are_whitespace",
       "edges_are_whitespace", "partitions_the_line"],
      pct=[("% partitioning", "partitions_the_line", "verses_with_spans")])
```

```python
# TEST 5. Reconstruction of the verse line from the word (slot) tier.
#
# Method: the slot type is `word`, so the reconstruction uses the text API applied to the verse.
# The format `text-trans-full` is defined as `{voc}{trailer}`, which concatenates each word with
# the material that follows it, and the result is compared to the `line` feature of the verse. The
# comparison is made on the whole string after trimming the trailing space that the final word
# contributes, so it is exact at the level of the character and includes every space.
# `letters_only` repeats the comparison under the letters class, and `utf8_spell_line` repeats the
# exact comparison in Aramaic square script, using `text-orig-full` against `line_utf8`.
#
# What it measures: whether the published slot sequence of a verse matches the published verse
# line. This is the end-to-end statement about the dataset, and the one a reader consuming the
# corpus through T.text experiences directly.
#
# Interpretation: the second table restricts the population to the verses in which every word
# carries morphological analysis, separating the verses the decomposition can be evaluated
# against from the verses where the source material supplies no analysis. The column
# `words_with_morphology` gives the same quantity at the level of the word.

test5 = defaultdict(Counter)
line_failures = []
for _k, _vns in verse_nodes.items():
    _r = test5[corpus_of_verse[_k]]
    line = max({F.line.v(v) or "" for v in _vns})
    line_utf8 = max({F.line_utf8.v(v) or "" for v in _vns})
    ws = [w for v in _vns for w in L.d(v, otype="word")]
    built = "".join((F.voc.v(w) or "") + (F.trailer.v(w) or "") for w in ws)
    built_utf8 = "".join((F.voc_utf8.v(w) or "") + (F.trailer.v(w) or "") for w in ws)
    analysed = all(F.source.v(w) == "word" for w in ws)
    ok = built.strip() == line
    _r["verses"] += 1
    _r["words"] += len(ws)
    _r["words_with_morphology"] += sum(1 for w in ws if F.source.v(w) == "word")
    _r["words_spell_line"] += ok
    _r["letters_only"] += letters(built) == letters(line)
    _r["utf8_spell_line"] += built_utf8.strip() == line_utf8
    _r["every_word_analysed"] += analysed
    _r["analysed_and_spelled"] += analysed and ok
    if not ok:
        line_failures.append((corpus_of_verse[_k], _k, line, built.strip()))

table("TEST 5  the words of a verse spell the verse line, character for character", test5,
      ["verses", "words_spell_line", "letters_only", "utf8_spell_line"],
      pct=[("% exact", "words_spell_line", "verses"), ("% letters", "letters_only", "verses")])
print()
table("TEST 5  restricted to the verses in which every word carries morphology", test5,
      ["verses", "words", "words_with_morphology", "every_word_analysed", "analysed_and_spelled"],
      pct=[("% words", "words_with_morphology", "words"),
           ("% of those verses", "analysed_and_spelled", "every_word_analysed")])
results["5 words compose the verse line"] = (sum(test5[c]["verses"] for c in CORPORA),
                                             sum(test5[c]["words_spell_line"] for c in CORPORA))
```

```python
# Repeats test 5 through T.text and classifies its failures, then shows the point at which two of
# them first diverge.
#
# The repetition through the text API checks the hand-written concatenation above. It covers the
# verses that occupy a single verse node, because T.text takes one node at a time.
#
# The classification identifies which tier is responsible for each failure. A verse with nothing
# published to spell it holds only words whose `source` is `none`, so the manuscript records a
# line that the source material never analysed and never transcribed at the word tier. A verse in
# which some word carries no morphological analysis is short by that word. A verse in which every
# word is analysed and the reconstruction still differs represents the residue attributable to the
# alignment.

_agree = _differ = 0
for _k, _vns in verse_nodes.items():
    if len(_vns) == 1:
        if T.text(_vns[0], fmt="text-trans-full").strip() == (F.line.v(_vns[0]) or ""):
            _agree += 1
        else:
            _differ += 1
print(f"T.text(verse, fmt='text-trans-full') == line:  {_agree:,} agree, {_differ:,} differ\n")

kinds = Counter()
residue = []
for _c, _k, _line, _built in line_failures:
    ws = [w for v in verse_nodes[_k] for w in L.d(v, otype="word")]
    if not _built:
        kinds["nothing published to spell it with"] += 1
    elif any(F.source.v(w) != "word" for w in ws):
        kinds["some word carries no morphology"] += 1
    else:
        kinds["every word analysed, residue of the alignment"] += 1
        residue.append((_c, _k, _line, _built))
for kind, n in kinds.most_common():
    print(f"  {n:>5,}  {kind}")


def first_divergence(a, b):
    i = next((i for i, (x, y) in enumerate(zip(a, b)) if x != y), min(len(a), len(b)))
    return i, a[max(0, i - 28):i + 28], b[max(0, i - 28):i + 28]


print()
for _c, _k, _line, _built in residue[:2]:
    at, la, lb = first_divergence(_line, _built)
    print(f"{_c} {_k[0]} uid {_k[1]}, first difference at character {at}")
    print(f"  line : ...{la}...")
    print(f"  words: ...{lb}...")
```

```python
# The word tier's offsets, reported here as coverage across all words, alongside the resolution
# rate for the words that carry one.
#
# A word's `start_idx` and `end_idx` are populated only where the line, sliced at those offsets,
# already returns the word's own text. Because of this, "the offsets resolve against the line" is
# true of every word that carries an offset by construction, and quoting that figure on its own
# would be quoting a tautology: 100% of a population defined as the part that passes. The number
# that carries information is how many words carry an offset at all, so both figures are printed
# together here, and the second should always be read alongside the first.
#
# The resolution rate is still worth checking, because the offset passes through several stages
# between where it is first computed and where it appears in this dataset. A fault introduced at
# any of those stages would show up here as a slice that no longer returns its word.
#
# `part` explains why a word may legitimately carry no offset. A word cut by an editorial boundary
# publishes one node per side, each holding only its own characters, so the population being
# measured is made up of word pieces, each smaller than the whole source word.

offsets = defaultdict(Counter)
for _k, _vns in verse_nodes.items():
    _r = offsets[corpus_of_verse[_k]]
    line = max({F.line.v(v) or "" for v in _vns})
    for _w in (w for v in _vns for w in L.d(v, otype="word")):
        voc = (F.voc.v(_w) or "").strip()
        _r["words"] += 1
        _r["with_surface"] += bool(voc)
        _r["split_pieces"] += (F.part.v(_w) or 1) > 1
        a, b = F.start_idx.v(_w), F.end_idx.v(_w)
        if a is None:
            _r["no_offset"] += 1
            continue
        _r["with_offset"] += 1
        _r["slice_is_word"] += line[a - 1:b] == voc

table("the word tier's offsets, as coverage", offsets,
      ["words", "with_surface", "split_pieces", "with_offset", "no_offset", "slice_is_word"],
      pct=[("% carrying an offset", "with_offset", "words"),
           ("% of those resolving", "slice_is_word", "with_offset")])
results["word offsets carried by the word tier"] = (
    sum(offsets[c]["words"] for c in CORPORA),
    sum(offsets[c]["with_offset"] for c in CORPORA))
results["word offsets that resolve, of those carried"] = (
    sum(offsets[c]["with_offset"] for c in CORPORA),
    sum(offsets[c]["slice_is_word"] for c in CORPORA))
```

```python
# TEST 6. Agreement of the stored ordinals with the publication order.
#
# Method: within each verse, read the segments and the word groups in node order, the order the
# dataset publishes, and count the adjacent pairs whose stored ordinal decreases. For segments the
# ordinals are `rank` and `start_idx`. For word groups the ordinal is the pair (`rank`, `reading`),
# because groups competing for one position share a rank and are ordered among themselves by
# reading. The final column tests whether the ranks of the word groups of a verse form a dense
# sequence starting from zero, the property that makes rank usable as a word index.
#
# What it measures: whether a consumer reading the corpus in node order and a consumer sorting by
# the stored ordinals obtain the same sequence. An inversion would allow a verse to be published in
# an order that its own features contradict, a fault the preceding tests would not detect, because
# each of them reads in node order throughout.

test6 = defaultdict(Counter)
for _k, _vns in verse_nodes.items():
    _r = test6[corpus_of_verse[_k]]
    _r["verses"] += 1
    segs = [s for v in _vns for s in L.d(v, otype="segment")]
    ranks = [F.rank.v(s) for s in segs]
    starts = [F.start_idx.v(s) for s in segs]
    _r["segments"] += len(segs)
    _r["segment_rank_inversions"] += sum(
        1 for a, b in zip(ranks, ranks[1:]) if a is not None and b is not None and b < a)
    _r["segment_span_inversions"] += sum(
        1 for a, b in zip(starts, starts[1:]) if a is not None and b is not None and b < a)
    gs = [(F.rank.v(g), F.reading.v(g)) for v in _vns for g in L.d(v, otype="word_group")]
    _r["word_groups"] += len(gs)
    _r["group_order_inversions"] += sum(1 for a, b in zip(gs, gs[1:]) if b < a)
    slots = {rank for rank, _ in gs}
    _r["slot_rank_dense"] += sorted(slots) == list(range(len(slots)))

table("TEST 6  stored ordinals against publication order", test6,
      ["verses", "segments", "segment_rank_inversions", "segment_span_inversions",
       "word_groups", "group_order_inversions", "slot_rank_dense"],
      pct=[("% dense", "slot_rank_dense", "verses")])
results["6 ordinals agree with node order"] = (
    sum(test6[c]["verses"] for c in CORPORA), sum(test6[c]["slot_rank_dense"] for c in CORPORA))
```

```python
# TEST 7. Case, spacing and stray diacritics in the ETCBC transliterated features.
#
# Method: count, per feature, the nodes whose value contains a lowercase letter. In the scheme
# the transliteration module defines, a lowercase letter is the final form of its consonant, so
# that `n` is a final nun against a medial `N`. The only lowercase letters that occur are
# `k m n p y`, and
# they are finals and nothing else. `y` is a final tsade, not a yod: yod is `J`.
#
# That is this repository extending ETCBC rather than following it. BHSA writes `MLK>` and
# `<BDJK`, all uppercase, because in Hebrew the final form is positionally predictable and so
# carries no information. Here the language is agglutinative and the decomposition of the source
# joins words, which is why a final letter can appear mid-string, and the
# reviewer who raised this read those 128 cases as errors. They are not errors; they are the
# reason the distinction cannot be re-derived from position.
#
# The distinction is nonetheless redundant HERE, which is what makes uppercasing safe. Every
# square-script column is derived directly from the source transliteration, selecting the final
# glyph from the
# own case; `ETCBC_TRANSLITERATION_HEBREW_CONSONANTS` is referenced only by its own definition
# and there is no ETCBC -> Hebrew path at all. So the final form is recorded twice, once in the
# square script and once in the case of a transliteration that claims to be ETCBC, and it is the
# transliteration that is wrong about itself. `upper()` is applied at the publication boundary,
# the stored columns keep the case of the source, and uppercasing preserves length so that the
# offsets still resolve.
#
# What it measures: whether the published transliteration is ETCBC. Every column must be 0.
#
# Two further integrity populations ride along. The internal space in `cons` USED TO BE
# deliberate. A lexical entry filed as several orthographic words occupied a single slot, and
# the space was restored so that the published line reconstructed character for character, which
# is why every other test here compares through `letters()`. The figure was asserted at its
# documented size of 2,750 rather than at zero, and the cell described splitting those runs as
# deferred work for which the figure was the starting point.
#
# That work has since been done, and the expectation is now 0. A multi-word surface is split on
# its own whitespace and each part takes its own slot, which affects 5,636 words. No published
# consonantal form holds an internal space, so this check is an invariant rather than a pinned
# ceiling: a space returning to the feature would mean that a surface had reached the slot tier
# unsplit. The ordinals that number those parts are not published, because they describe the
# internal segmentation model rather than the text, so this check is the published evidence that
# the split occurred.
# U+0740 is the Genizah lacuna mark, and zero is the wrong expectation for it. This is a
# correction to the first version of this cell. Measured over the published corpus there are 9
# occurrences in `voc` and 94 in `g_voc`, and every one of the 9 is a word whose ENTIRE `voc` is a
# run of three to seven dots, with no `id`, no `cons` and no offset: a lacuna standing in for a
# word, which is the one place the mark belongs. None of the 9 is on a word the morphology was
# placed on. On the group tier 3 of the 94 are mixed with letters, which is a partially lacunose
# orthographic word and is equally correct. So the invariant is not absence but segregation: the
# mark must never share a word with a letter, and the published line must hold none at all. It
# holds none,
# which is exactly why those 9 words are correctly unlocated.

# `lex0` is `lex` with the BHSA disambiguation material removed, so it is an ETCBC string in its
# own right and has to fold with the rest. It is named conditionally because this cell has to run
# against a dataset compiled before the feature existed as well as after.
ETCBC_WORD = ("cons", "voc", "lex", "root") + (
    ("lex0",) if hasattr(F, "lex0") else ())
ETCBC_GROUP = ("g_cons", "g_voc")
_LOWER = re.compile(r"[a-z]")
DOT = "\u0740"


def lowercase_in(s):
    return bool(s) and bool(_LOWER.search(s))


test7 = defaultdict(Counter)
for _k, _vns in verse_nodes.items():
    _r = test7[corpus_of_verse[_k]]
    for _v in _vns:
        _r["verses"] += 1
        _r["line"] += lowercase_in(F.line.v(_v))
        _r["line_u0740"] += DOT in (F.line.v(_v) or "")
        for _s in L.d(_v, otype="segment"):
            _r["segments"] += 1
            _r["segment"] += lowercase_in(F.segment.v(_s))
        for _g in L.d(_v, otype="word_group"):
            _r["word_groups"] += 1
            for _f in ETCBC_GROUP:
                _r[_f] += lowercase_in(Fs(_f).v(_g))
            _gv = F.g_voc.v(_g) or ""
            _r["g_voc_u0740"] += DOT in _gv
            _r["g_voc_u0740_mixed"] += DOT in _gv and bool(_gv.strip(DOT + " "))
        for _w in L.d(_v, otype="word"):
            _r["words"] += 1
            _clean = True
            for _f in ETCBC_WORD:
                _val = Fs(_f).v(_w)
                _bad = lowercase_in(_val)
                _clean &= not _bad
                _r[_f] += _bad
                _r[f"{_f}_mid"] += lowercase_in(_val[:-1] if _val else None)
            _r["words_upper"] += _clean
            _r["cons_space"] += " " in (F.cons.v(_w) or "").strip()
            _vv = F.voc.v(_w) or ""
            _r["voc_u0740"] += DOT in _vv
            # The two ways the mark could be wrong: sharing a word with a letter, or sitting on a
            # word the morphology was placed on rather than on a stand-in for a missing one.
            _r["voc_u0740_mixed"] += DOT in _vv and bool(_vv.strip(DOT + " "))
            _r["voc_u0740_on_a_word"] += DOT in _vv and F.source.v(_w) == "word"

table("TEST 7  lowercase in the ETCBC features, by tier", test7,
      ["words", "cons", "voc", "lex", "root", "word_groups", "g_cons", "g_voc",
       "verses", "line", "segments", "segment"])
print()
table("TEST 7  a final letter written mid-string, and the two diacritic populations", test7,
      ["words_upper", "cons_mid", "voc_mid", "lex_mid", "root_mid",
       "cons_space", "voc_u0740", "voc_u0740_mixed", "voc_u0740_on_a_word",
       "g_voc_u0740", "g_voc_u0740_mixed", "line_u0740"])
print()

_t7 = Counter()
for _c in CORPORA:
    _t7.update(test7[_c])
for _f in (*ETCBC_WORD, *ETCBC_GROUP, "line", "segment"):
    check(f"lowercase in `{_f}`", _t7[_f], 0,
          "lowercase is the final form; the ETCBC features are uppercased at the "
          "publication boundary and the final glyph stays in the *_utf8 columns")
for _f in ETCBC_WORD:
    check(f"a final letter mid-string in `{_f}`", _t7[f"{_f}_mid"], 0)
# Segregation, not absence. The 9 and the 94 are pinned as ceilings, the way `cons_space` is.
check("U+0740 sharing a word with a letter", _t7["voc_u0740_mixed"], 0,
      "a lacuna mark on a word that has letters of its own is a stray diacritic; a word "
      "that is nothing but lacuna marks is a lacuna, and that is all 9 of them")
check("U+0740 on a word the morphology reached", _t7["voc_u0740_on_a_word"], 0)
check("U+0740 in `line`", _t7["line_u0740"], 0,
      "the published line carries no lacuna mark, which is why the 9 lacuna words "
      "carry no offset into it")
check("U+0740 in `voc`", _t7["voc_u0740"], 9,
      "DOCUMENTED: 9 words whose whole surface is a lacuna, all in neofiti_full")
check("U+0740 in `g_voc`", _t7["g_voc_u0740"], 94,
      "DOCUMENTED: 89 neofiti_full and 5 base_raw groups, 3 of them a lacuna beside letters")
check("internal space in `cons`", _t7["cons_space"], 0,
      "the multi-word surfaces are split on their own whitespace, so each part now takes its "
      "own slot. An invariant since the rebuild, where it had been a pinned ceiling of 2,750. "
      "A space returning to this feature means a surface reached the slot tier unsplit")

results["7 the ETCBC features are uppercase"] = (
    sum(test7[c]["words"] for c in CORPORA), sum(test7[c]["words_upper"] for c in CORPORA))
```

```python
# TEST 8. The lexeme inventory of each corpus, against the inventory of the corpora whose `lex`
# is drawn from a lemma.
#
# Method: collect the distinct lexemes per corpus, then count how many of them occur in neither
# Neofiti corpus. Those two are the reference set because they are the only two whose `lex` the
# view layer sourced from the lemma tree, while the other three read the surface column of the
# analysed unit. All five corpora are analysed against one dictionary, so a lexeme attested in
# Pseudo-Jonathan and in no Palestinian Targum is either genuine Late Jewish Literary Aramaic
# vocabulary or an artefact, and the proportion tells which.
#
# What it measures: whether `lex` is a lexeme at all. A surface form inflates the inventory
# without limit, because `>JTJHJBT`, `D>JTJHJBT` and `W>JTJHJBT` are three values where the lemma
# `JHB` is one, and almost none of them can recur in another corpus by chance. `lex_eq_cons` is
# the same fact read a second way: a lexeme equals the surface form only where the word is
# uninflected and unprefixed, so a high rate at four letters or more is the signature of a
# surface-form lexeme, and the verb column is the sharpest form of it, because an Aramaic verb
# is
# practically never its own dictionary headword.
#
# Interpretation. The inventory figures are now pinned, because they have been measured off the
# regenerated view bodies rather than predicted: 2,484 / 2,281 / 2,946 / 1,824 / 1,183 distinct
# lexemes, 2,500 in the reference union and 3,692 in the corpus as a whole, against 20,779
# before. The three corpora that used to publish a surface form fall from 14,104 / 5,794 / 2,970
# values outside the reference inventory to 1,016, 176 and 149, so 90.9 per cent of what looked
# like Pseudo-Jonathan vocabulary was an artefact of the source column, and what is left is the
# right order of magnitude for genuine Late Jewish Literary Aramaic.
#
# Set sizes are pinned and node counts are not. A distinct-value count is the same whether it is
# taken over matview rows or over TF nodes, because `generate_xml` merges only words whose whole
# attribute dictionary is identical, the lexeme included. The two word counts count nodes,
# so they are reported and left to the `results` row.
#
# `lex_eq_cons` stops being a defect measure here and the columns are kept for continuity. With a
# surface-form lexeme the equality was itself the defect, whereas with a lemma it is ordinary,
# since an unprefixed singular noun is its own headword. The figure therefore rises even though
# the fault is gone. The
# column that still means something is the verb one: 14,233 verbs equalled their own surface form
# before and 93 do after, because an Aramaic verb is practically never its dictionary headword.
#
# It reads `lex0` where the dataset has one. Following BHSA, `lex` carries the disambiguation
# material, which is `[` for a verb, `/` for a noun, and zero or more `=` signs for lexemes
# sharing consonants. `lex0` is the same value with that material removed. The inventory question
# is about the lexeme, so it is asked of the undisambiguated form.

LEX0 = "lex0" if hasattr(F, "lex0") else "lex"
print(f"reading the lexeme from `{LEX0}`"
      + ("" if LEX0 == "lex0" else "  (no `lex0` in this dataset yet)"))

test8 = defaultdict(Counter)
lex_sets = defaultdict(set)
_lexf = Fs(LEX0)
for _k, _vns in verse_nodes.items():
    _c = corpus_of_verse[_k]
    _r = test8[_c]
    for _v in _vns:
        for _w in L.d(_v, otype="word"):
            _lex, _cons = _lexf.v(_w), F.cons.v(_w)
            _r["words"] += 1
            _r["lex_present"] += _lex is not None
            if _lex is None:
                continue
            lex_sets[_c].add(_lex)
            _r["lex_eq_cons"] += _lex == _cons
            if _cons is not None and len(_cons) >= 4 and _lex == _cons:
                _r["eq_4plus"] += 1
                _r["eq_4plus_verb"] += F.sp.v(_w) == "verb"

_reference = lex_sets["neofiti_base_raw"] | lex_sets["neofiti_full"]
for _c in CORPORA:
    test8[_c]["distinct_lex"] = len(lex_sets[_c])
    test8[_c]["distinct_outside"] = len(lex_sets[_c] - _reference)

for _k, _vns in verse_nodes.items():
    _r = test8[corpus_of_verse[_k]]
    for _v in _vns:
        for _w in L.d(_v, otype="word"):
            _lex = _lexf.v(_w)
            if _lex is not None and _lex not in _reference:
                _r["words_outside"] += 1

table("TEST 8  the lexeme inventory, against the lemma-sourced corpora", test8,
      ["words", "lex_present", "distinct_lex", "distinct_outside", "words_outside"],
      pct=[("% of the inventory outside", "distinct_outside", "distinct_lex"),
           ("% of words", "words_outside", "lex_present")])
print()
table("TEST 8  the lexeme that is its own surface form", test8,
      ["lex_present", "lex_eq_cons", "eq_4plus", "eq_4plus_verb"],
      pct=[("% lex == cons", "lex_eq_cons", "lex_present")])
print()
print(f"  {len(_reference):,} distinct lexemes in the reference inventory; "
      f"{len(set().union(*lex_sets.values())):,} in the corpus as a whole")
print()

# The inventory is pinned. It was measured against the regenerated view bodies before the
# materialised views were rebuilt, so these figures are a prediction that the rebuild has to
# reproduce rather than a
# description of whatever the dataset happens to hold.
# It was re-pinned after three correction campaigns. The inventory grew because the lemmas
# spelled with a tet are now spelled with one, rising from 2,484 to 2,520 in the full Neofiti
# view, and shrank
# because a lemma the tree used to truncate is now the headword the inventory already held
# already held. Both corrections move the same column, so the net change per corpus was small,
# from 3,692 to 3,771.
#
# Re-pinned again against the 496,532-slot build, and the cause is the onomasticon. The
# corpus-wide figure moved from 3,771 to 4,728, and 963 of the 4,728 are lexemes that occur only
# on a word classified as a proper name. That campaign gave 29,384 words a name category and
# gave the proper names a lexeme they had not previously had, so a divine name, a personal name
# and a place name now each resolve to an entry rather than to nothing. Net of those the
# inventory stands at 3,765 against the previous 3,771, which measures how little else moved.
#
# These figures were confirmed against the database before being written here, which is what
# distinguishes a re-pin from a rubber stamp: `count(DISTINCT word_lex0)` over the five `*_tf`
# views returns 1,300 / 2,125 / 2,931 / 3,163 / 3,682 and 4,728 corpus-wide, identical to the
# TF figures below on every corpus. The prediction mechanism worked; what it predicted was a
# pre-Phase-2 corpus. `sql/checks/baseline.sql` 15b and 15c read the same column off the views.
_EXPECTED_DISTINCT = {"neofiti_full": 3163, "neofiti_base_raw": 2931, "psj": 3682,
                      "frag_targ": 2125, "cairo_genizah": 1300}
_EXPECTED_OUTSIDE = {"neofiti_full": 0, "neofiti_base_raw": 0, "psj": 1285,
                     "frag_targ": 274, "cairo_genizah": 173}
if LEX0 == "lex0":
    for _c in CORPORA:
        check(f"distinct lexemes in {_c}", len(lex_sets[_c]), _EXPECTED_DISTINCT[_c])
    for _c in CORPORA:
        check(f"lexemes {_c} shares with neither Neofiti corpus",
              len(lex_sets[_c] - _reference), _EXPECTED_OUTSIDE[_c],
              "a lexeme in one corpus and in neither lemma-sourced corpus is either "
              "genuine LJLA vocabulary or an artefact of the source column")
    check("distinct lexemes, corpus-wide", len(set().union(*lex_sets.values())), 4728,
          "20,779 before the single-source correction and 3,771 before the onomasticon, "
          "against roughly 9,230 for the whole Hebrew Bible. 963 of these occur only on a "
          "word classified as a proper name")
    check("lexemes shared by all five corpora",
          len(set.intersection(*(lex_sets[c] for c in CORPORA))), 939,
          "877 before the onomasticon. A proper name attested in all five manuscripts now "
          "lexeme across them instead of carrying none in any")
else:
    print("  (pinned inventory figures skipped: this dataset has no `lex0`)")
print()

results["8 the lexeme is drawn from one lexicon"] = (
    sum(test8[c]["lex_present"] for c in CORPORA),
    sum(test8[c]["lex_present"] - test8[c]["words_outside"] for c in CORPORA))
```

```python
# TEST 9. Two features that carry something other than what they are named for.
#
# `gloss` is non-null on every word in the corpus, which is not a fact about lexical coverage.
# The
# tail of its COALESCE in the view layer is `concat_ws(', ', category, description)`, the same
# expression published beside it as `description`, so where the source has no gloss of any kind
# and no full definition, `gloss` is handed the lexical category, which is the literal value
# "nominal, proper name". A consumer filtering on a non-null gloss therefore receives the whole
# corpus.
# `gloss_is_a_category` names that population.
#
# `text` is two features under one name. The word element's character data is set from
# `cons_utf8` and the XML converter promotes element text to a feature called `text`, while the
# schema already carries `text` for the manuscript name, and `text` is a section feature
# (`@sectionFeatures=text,book,chapter`), so the collision is with the corpus's own spine. The two
# halves never disagree: across every word that has one, `text` differs from `cons_utf8` in zero
# cases, and that is the proof that the word half carries no information and can simply stop
# being emitted. This test asserts both the redundancy and its removal, so it keeps its value
# after the fix: if `text` ever comes back on a word carrying something of its own, the first
# check fails rather than the second.

test9 = defaultdict(Counter)
for _k, _vns in verse_nodes.items():
    _r = test9[corpus_of_verse[_k]]
    for _v in _vns:
        for _w in L.d(_v, otype="word"):
            _g, _d, _t = F.gloss.v(_w), F.description.v(_w), F.text.v(_w)
            _r["words"] += 1
            _r["gloss_present"] += _g is not None
            _r["gloss_is_a_category"] += _g is not None and _g == _d
            _r["has_definition"] += F.definition.v(_w) is not None
            _r["carries_text"] += _t is not None
            _r["text_differs"] += _t is not None and _t != F.cons_utf8.v(_w)

table("TEST 9  gloss against description, and the word half of `text`", test9,
      ["words", "gloss_present", "gloss_is_a_category", "has_definition",
       "carries_text", "text_differs"],
      pct=[("% gloss is a category", "gloss_is_a_category", "gloss_present")])
print()

_t9 = Counter()
for _c in CORPORA:
    _t9.update(test9[_c])
check("`text` on a word differing from `cons_utf8`", _t9["text_differs"], 0,
      "the word half of `text` is a byte-for-byte copy of cons_utf8; if this is ever "
      "non-zero it has started carrying information and must not simply be dropped")
check("word nodes carrying `text`", _t9["carries_text"], 0,
      "`text` is also the section feature of the `text` node type; the word half is "
      "redundant and is no longer emitted")
check("words whose `gloss` is a lexical category", _t9["gloss_is_a_category"], 0,
      "the COALESCE tail hands `gloss` the value published beside it as `description`")

results["9 `text` is one feature"] = (
    sum(test9[c]["words"] for c in CORPORA),
    sum(test9[c]["words"] - test9[c]["carries_text"] for c in CORPORA))
```

```python
# TEST 10. The section census, and the two places the corpus admits it does not know where it is.
#
# The method is to walk the section tiers, which are `book`, `chapter` and `verse`,
# and count the chapter nodes whose `chapter` is the string "0", the verse nodes carrying no
# `verse`, and the material each of those holds.
#
# Chapter 0 means "no chapter could be derived". `src/main/domain/subtext.py:40-67` initialises
# `chapter` to 0 and returns it when the scraped subtext name matches none of its three regexes;
# the view layer then keeps the 0 on purpose, because Text-Fabric requires a value and not a null
# on a section feature, and a verse whose section is absent cannot be addressed at all. Both
# cases are unlocalised fragments, namely a leaf of `fragment_targum_p` and one of
# `fragment_targum_v`
# whose heading names the book but no chapter.
#
# This has been restated. The verses with no verse number were the same problem one tier down,
# and this
# cell pinned them at NINE, "confined to the same two manuscripts". Both halves of that have
# changed, and in opposite directions:
#
#   The count of verse nodes carrying no verse number is now zero, and that follows from the
#   section model rather than from a repair. Making the verse a section level is what allows a
#   section address to resolve, and a section address may not be null, so the null became a
#   SENTINEL: 0, or 0 and -2 where one chapter holds two of them, since two verses may not
#   share one address. Nothing was localised; the unknown is now written down instead of
#   left absent, which is why the check below is an invariant and the sentinel count beside
#   it is the documented ceiling that replaces the old 9.
#
#   The sentinel steps by TWO, and that is the one thing here that is not about this corpus.
#   context-fabric compiles an int feature into an array whose own missing-value sentinel IS
#   -1 (`cfabric.io.compiler._check_sentinel_collision` says so and warns), so a published -1
#   loads back as None. Measured on the build before this one: node 566613,
#   `fragment_targum_v` at Deuteronomy 28, carried `verse` -1, `text-fabric` read it -1 and
#   the reading client returned a null, and it was the only such value in any integer feature.
#   `-1` is therefore reserved, the sentinel descends 0, -2, -4, and the last check in this
#   cell is the guard. That check reads the shipped `.tf` FILES and not the loaded API,
#   because under cfabric a published -1 arrives as None and asking the API whether anything
#   carries -1 would answer no on exactly the fault it exists to catch.
#
#   The population is 11, not 9, and it spans THREE manuscripts rather than two. Nine are in
#   `fragment_targum_p` and `fragment_targum_v` as the old note said; the other two are
#   Cairo Genizah fragments, `_g` at Exodus 15 and `_t` at Deuteronomy 34, and they are the
#   two largest of the eleven at 42 and 148 words. `config/xml.yml` still described this
#   population as "nine `frag_targ` verses" and is corrected with this cell.
#
# What it measures: the size of the undetermined population, asserted as a documented CEILING.
# This test PASSES today. It is here as the regression guard: an unlocalised fragment is a
# property of the source, so the figure may legitimately be explained, but it must never grow
# silently, and zero is not the target, because a chapter 0 that became chapter 1 would be a
# fabrication.
#
# `chapter` and `verse` are published with `valueType=str` and are absent from `config/xml.yml`'s
# `intFeatures`, so a consumer sorting the 51 distinct chapter values lexically puts "10" before
# "2". The last two checks are that contract, and they fail until those two names are added.

test10 = defaultdict(Counter)
for _t in F.otype.s("text"):
    _r = test10[corpus_of(_t)]
    _r["texts"] += 1
    for _b in L.d(_t, otype="book"):
        _r["books"] += 1
    for _c in L.d(_t, otype="chapter"):
        _r["chapters"] += 1
        if F.chapter.v(_c) in ("0", 0):
            _r["chapter_0"] += 1
            _r["verses_in_chapter_0"] += len(L.d(_c, otype="verse"))
            _r["words_in_chapter_0"] += len(L.d(_c, otype="word"))
    for _v in L.d(_t, otype="verse"):
        _r["verses"] += 1
        _ch = L.u(_v, otype="chapter")
        _located = F.verse.v(_v) is not None and not (
            _ch and F.chapter.v(_ch[0]) in ("0", 0))
        _r["located"] += _located
        if F.verse.v(_v) is None:
            _r["no_verse"] += 1
            _r["words_in_no_verse"] += len(L.d(_v, otype="word"))

table("TEST 10  the section census", test10,
      ["texts", "books", "chapters", "verses", "chapter_0", "verses_in_chapter_0",
       "words_in_chapter_0", "no_verse", "words_in_no_verse", "located"],
      pct=[("% located", "located", "verses")])
print()
for _t in F.otype.s("chapter"):
    if F.chapter.v(_t) in ("0", 0):
        _b, _x = L.u(_t, otype="book"), L.u(_t, otype="text")
        print(f"  chapter 0: {F.text.v(_x[0])} {F.book.v(_b[0])}, "
              f"{len(L.d(_t, otype='verse'))} verse, {len(L.d(_t, otype='word'))} words")
print()

_t10 = Counter()
for _c in CORPORA:
    _t10.update(test10[_c])
check("chapter nodes at chapter 0", _t10["chapter_0"], 2,
      "an unlocalised fragment whose heading names the book but no chapter; TF needs a "
      "value and not a null on a section feature, so the 0 is kept deliberately")
check("words under chapter 0", _t10["words_in_chapter_0"], 13)
check("verse nodes carrying no `verse`", _t10["no_verse"], 0,
      "the verse is a section level and a section address cannot be null, so the nine "
      "nulls became sentinels. An invariant since: a verse node with no `verse` is one that "
      "cannot be addressed at all")
# The sentinel population replaces the old count of 9 as the documented ceiling. Asserted
# here rather than merely reported, because it is the figure that must never grow silently. A
# verse to which the source gives no number is a property of that source, and inventing one would
# be a
# fabrication exactly as a chapter 0 becoming chapter 1 would be.
def _verse_number(v):
    """`verse` as an int, with a missing value folded in as a sentinel rather than raised on.

    The check above asserts that no verse node is missing `verse`, so this cannot mask a null in
    the data. It exists because cfabric returns None for a published -1, and a bare
    converting the value then raises in the middle of a census, which is how the
    collision documented above was found.
    """
    n = F.verse.v(v)
    return 0 if n is None else int(n)


_sentinel = [_v for _v in F.otype.s("verse") if _verse_number(_v) <= 0]
check("verse nodes publishing a sentinel verse number", len(_sentinel), 11,
      "0, or 0 and -2 where one chapter holds two: nine in the two fragment_targum manuscripts "
      "and two in the Cairo Genizah fragments `_g` (Exodus 15) and `_t` (Deuteronomy 34)")
print(f"  sentinel verse numbers published: "
      f"{dict(Counter(sorted(_verse_number(_v) for _v in _sentinel)))}")


def tf_value_occurrences(feature, wanted):
    """Count occurrences of one literal value in a shipped `.tf` node feature file.

    Read from the file, not through the API. A `.tf` node feature line is `<spec>\t<value>` or a
    bare `<value>` continuing the previous spec, after an `@`-header terminated by the first line
    that does not start with `@`.
    """
    path = TF_DIR / f"{feature}.tf"
    if not path.exists():
        return None
    n, in_header = 0, True
    with path.open(encoding="utf8") as fh:
        for line in fh:
            if in_header:
                if line.startswith("@"):
                    continue
                in_header = False
                continue                      # the blank line that closes the header
            if line.rstrip("\n").rsplit("\t", 1)[-1] == wanted:
                n += 1
    return n


INT_FEATURES = yaml.safe_load((REPO_ROOT / "config/xml.yml").read_text())["intFeatures"]
_reserved = {_f: tf_value_occurrences(_f, "-1") for _f in INT_FEATURES}
_reserved = {_f: _n for _f, _n in _reserved.items() if _n}
check("int-feature occurrences carrying the reserved -1", sum(_reserved.values()), 0,
      "-1 is context-fabric's missing-value sentinel for an int feature, so a published -1 "
      "loads back as None and `F.<f>.v(n) <= 0` raises instead of answering: "
      + (", ".join(f"{_f}={_n}" for _f, _n in _reserved.items()) or "none"))
print(f"  int features scanned for the reserved -1: {len(INT_FEATURES)}")
check("chapter nodes in total", _t10["chapters"], 1038,
      "the section tier itself; a change here is a change in what the corpus publishes")

# The manuscript tier, and the two numbers that must not be confused with each other.
#
# `text` names the manuscript everywhere EXCEPT Neofiti, where `neofiti_base` and `neofiti_full`
# are two views of one codex, so the corpus has 46 `text` values over 45 manuscripts, and
# `manuscript` is the feature that states it. It is the one fact in the parallel witness relation
# that nothing else published can supply: two verses are parallel iff they share
# `(book, chapter, verse)`, which IS derivable, but telling a parallel (two manuscripts translate
# this verse) from a view (one manuscript under two editorial filters) is not. Derived without it
# the relation splits 49,384 witness / 20 view against the true 37,658 / 11,654.
#
# `text` node count is 46 rather than 68 because the tier is resumed. The serialiser writes one
# XML
# folder per verse-id RANGE, and six manuscripts span several, so the walk used to open a node per
# folder. Both numbers are asserted, because the failure mode they guard against is each other.
_texts = list(F.otype.s("text"))
_manuscripts = {F.manuscript.v(_t) for _t in _texts}
_names = {F.text.v(_t) for _t in _texts}
check("`text` nodes", len(_texts), 46,
      "one per manuscript name rather than one per directory, which was 68 before the tier "
      "was resumed")
check("distinct `text` names", len(_names), 46,
      "one node each, so iterating `F.otype.s('text')` no longer over-counts a manuscript")
check("distinct manuscripts", len(_manuscripts), 45,
      "46 names, 45 manuscripts; the single collapse is Neofiti 1, whose two corpora are two "
      "view types of one codex")
check("text nodes with no `manuscript`", sum(1 for _t in _texts if F.manuscript.v(_t) is None), 0)
_neofiti = {F.manuscript.v(_t) for _t in _texts if F.text.v(_t).startswith("neofiti")}
check("the Neofiti views that resolve to one codex", len(_neofiti), 1,
      "`neofiti_base` and `neofiti_full` must share one `manuscript` value, which is what makes "
      "a witness count right")
for _f in ("chapter", "verse"):
    _vt = tf_meta(_f).get("valueType")
    _n = len({Fs(_f).v(_n) for _n in F.otype.s(_f)} - {None})
    print(f"  {'ok  ' if _vt == 'int' else 'FAIL'}  {('`' + _f + '` valueType'):<46} "
          f"{_vt:>9}" + ("" if _vt == "int" else "  expected int"))
    if _vt != "int":
        violations.append((f"`{_f}` valueType", _vt, "int",
                           f"absent from config/xml.yml intFeatures, so a consumer sorting "
                           f"the {_n} distinct `{_f}` values gets them lexically, "
                           '"10" before "2"'))

results["10 sections are determined"] = (
    sum(test10[c]["verses"] for c in CORPORA), sum(test10[c]["located"] for c in CORPORA))
```

```python
# TEST 11. Every node feature counted by the node type its occurrences actually fall on.
#
# The method is to read the backing store of each node feature directly, which is a mapping
# keyed by node number, and to tally the node type of every key. That makes the matrix exact and
# inexpensive,
# over all 495,216 slots and every higher node, rather than sampled.
#
# What it measures: whether a feature name means ONE thing. Cell 3 above asks a version of this
# question from the first 400 nodes of each type, which describes the dataset well enough to read
# the tests by, but cannot see a feature that lives overwhelmingly on one type and occasionally on
# another. That is exactly the shape of the fault the review found: `text` is the manuscript name
# on 68 `text` nodes and the surface form on 490,496 words, and it is also a SECTION feature, so
# the collision is with the corpus's own addressing. Nothing in the serialisation chain could have
# caught it, because the word half is produced by the XML converter promoting element character
# data to a feature and therefore has no row in the published schema and no source column. It is
# the one feature
# in the dataset outside the contract the `views_tf` docstring claims is total.
#
# Some names legitimately measure one quantity at two tiers, and the assertion is that the
# overloaded set is EXACTLY those: `id` is an identifier, the same kind of thing wherever it
# appears, and `rank`, `start_idx` and `end_idx` are the ordinal and the two offsets, defined the
# same way for a word, a word group and a segment. `text` was a sixth, and the only one whose two
# readings were different kinds of thing. This is the generic guard: it fails on the NEXT such
# feature without anyone having to think of it.
#
# The two groups are now named apart, because they are permitted for
# different reasons:
#
#   `lex`, `lex0` and `lex_utf8` are overloaded deliberately. The lexeme node type is additive,
#   in that it carries the lexical data that was previously repeated on every token,
#   and BHSA likewise carries `lex` on `word` and on its own `lex` node type. The same string
#   at both tiers is the same kind of thing, which is precisely the test this cell applies, so
#   they belong in the allowed set rather than in an exemption.
#
#   `uid` has left that group, because the verse identifier is withheld. Such an identifier
#   collides almost
#   completely across the three family tables, so a consumer joining on it read the wrong row
#   silently. It is now a word feature only, and the second check below would have read that
#   as a tier that stopped carrying it. The check is therefore split, so that the tier pairs
#   must
#   still be paired, and `uid` is asserted to be single-tier, so re-publishing it at the verse
#   tier fails here rather than passing quietly.

TIER_PAIRS = {"id", "rank", "start_idx", "end_idx"}
LEXEME_TIER = {"lex", "lex0", "lex_utf8"}
ALLOWED_OVERLOAD = TIER_PAIRS | LEXEME_TIER

_types = list(F.otype.all)
matrix = {}
for _f in sorted(api.Fall()):
    if _f == "otype":
        continue
    _c = Counter()
    for _n in Fs(_f).data:
        _c[F.otype.v(_n)] += 1
    matrix[_f] = _c

_w0 = max(len("feature"), max(len(f) for f in matrix))
_w = {t: max(len(t), 9) for t in _types}
print("TEST 11  node features by the node type they occur on")
print("  " + "feature".ljust(_w0) + "  " + "  ".join(t.rjust(_w[t]) for t in _types)
      + "  " + "types".rjust(5))
print("  " + "-" * _w0 + "  " + "  ".join("-" * _w[t] for t in _types) + "  " + "-" * 5)
for _f, _c in matrix.items():
    _mark = "" if len(_c) == 1 else ("  <- tier pair" if _f in ALLOWED_OVERLOAD
                                     else "  <- OVERLOADED")
    print("  " + _f.ljust(_w0) + "  "
          + "  ".join((f"{_c[t]:,}" if _c[t] else "-").rjust(_w[t]) for t in _types)
          + f"  {len(_c):>5}" + _mark)
print()

overloaded = {f for f, c in matrix.items() if len(c) > 1}
print(f"  overloaded: {', '.join(sorted(overloaded)) or 'none'}")
print(f"  allowed:    {', '.join(sorted(ALLOWED_OVERLOAD))}")
check("features overloaded across node types", len(overloaded - ALLOWED_OVERLOAD), 0,
      "a name carrying two different kinds of thing: "
      + (", ".join(sorted(overloaded - ALLOWED_OVERLOAD)) or "none"))
check("tier-pair features that lost their second tier",
      len(ALLOWED_OVERLOAD - overloaded), 0,
      "`id`, `rank`, `start_idx` and `end_idx` are each defined at more than one tier, and "
      "`lex`, `lex0` and `lex_utf8` are published on `word` and on `lexeme`, and one of them "
      "falling "
      "back to a single type means a tier stopped carrying it")
check("`uid` tiers, after the verse identifier was withheld", len(matrix["uid"]), 1,
      "a word feature only. Unit identifiers collide across the three family tables, so a "
      "consumer joining on `verse/@uid` read the wrong row silently")

results["11 a feature name means one thing"] = (
    len(matrix), sum(1 for f, c in matrix.items() if len(c) == 1 or f in ALLOWED_OVERLOAD))
```

```python
# TEST 12. The features added in this version, and whether they say what they claim.
#
# The method is to assert that each feature sits on the tier the documentation places it on, and
# then to assert the property that makes it worth publishing. A feature that exists and is wrong
# is worse than one that is absent, because a consumer has no means of telling. A feature read
# from the wrong tier is worse still, because it returns None without raising.
#
# This version publishes six features that the previous release did not. Four of them sit on the
# lexeme tier, which is itself new, and one on the manuscript tier. The sixth is the parallel
# witness edge, which test 14 covers in full.
#
# A larger set of candidates was considered and withheld, and the assertions below require each
# of them to be absent. Three classes account for all of them. The first is identifiers of the
# relational source, whose values a consumer cannot resolve. The second is values recoverable
# from the rest of the dataset. The third is values that do not vary, which answer no question
# while occupying space on every node of their tier. Asserting absence matters as much as
# asserting presence, because a withdrawn feature that returns in a later rebuild would be a
# silent change to a citable dataset.

_WORD_TIER = ["lex0", "lex_utf8"]
_LEXEME_TIER = ["lex0", "lex_utf8", "freq_lex", "rank_lex"]
_TEXT_TIER = ["manuscript"]

# Withheld deliberately. `lex_id` and `lex_key` were identifiers, `language` and `dialect` did not
# vary usefully, `lex0` already recovers what `lex_dialects`, `gloss_norm` and `voc_lex` were
# consulted for at the lexeme tier, and `variant`, `nametype`, `piece` and `pieces` were either
# a single legend sentence or an ordinal of the internal segmentation model.
_WITHDRAWN = ["lex_id", "lex_key", "language", "dialect", "lex_dialects", "gloss_norm",
              "voc_lex", "variant", "nametype", "piece", "pieces"]

_all_features = TF_FEATURES


def _occurs_on(feat, otype):
    """How many nodes of `otype` carry `feat`.

    The result is 0 both when the feature is absent from the dataset and when it is present on
    another tier, which is the confusion this cell exists to detect.
    """
    if feat not in _all_features:
        return 0
    _f = Fs(feat)
    return sum(1 for _n in F.otype.s(otype) if _f.v(_n) is not None)


print("TEST 12  the features of this version, and the tier each one occupies")
_tiers = [("word", _WORD_TIER), ("lexeme", _LEXEME_TIER), ("text", _TEXT_TIER)]
for _otype, _feats in _tiers:
    for _f in _feats:
        _n = _occurs_on(_f, _otype)
        print(f"  {'ok  ' if _n else 'FAIL'}  {_f:<13} on {_otype:<7} {_n:>8,} nodes")
        check(f"`{_f}` occurs on the `{_otype}` tier", int(_n > 0), 1,
              "declared in config/xml.yml and emitted by tf_pipeline.SCHEMA_ROWS")
print()

print("TEST 12  the features withheld from this version")
for _f in _WITHDRAWN:
    _absent = _f not in _all_features
    print(f"  {'ok  ' if _absent else 'FAIL'}  {_f:<13} is withheld and absent")
    check(f"`{_f}` is absent from the dataset", int(not _absent), 0,
          "withheld as an unresolvable identifier, a recoverable value or a constant")
print()

# The word to lexeme relation. The lexeme tier is resumed and is not contained by the text tree,
# so a lexeme node's slots are distributed across the corpus and containment is the only route
# from a word to the properties of its lexeme.
lexeme_of = {}
for _lx in F.otype.s("lexeme"):
    for _w in L.d(_lx, otype="word"):
        lexeme_of[_w] = _lx

_MARKERS = set("[/=")
test12 = defaultdict(Counter)
for _k, _vns in verse_nodes.items():
    _r = test12[corpus_of_verse[_k]]
    for _v in _vns:
        for _w in L.d(_v, otype="word"):
            _r["words"] += 1
            _lex, _lex0 = F.lex.v(_w), F.lex0.v(_w)
            _lx = lexeme_of.get(_w)
            _r["lex_utf8"] += F.lex_utf8.v(_w) is not None
            _r["lexeme_node"] += _lx is not None
            if _lex0 is None and _lex is None:
                _r["no_lexeme"] += 1
                continue
            _r["lexeme"] += 1
            # The disambiguation shape, asserted per word rather than in aggregate. This single
            # predicate catches a marker placed on the wrong base, a base that has drifted away
            # from the lexeme it belongs to, and a stray character in either.
            _ok = (_lex is not None and _lex0 is not None
                   and _lex.startswith(_lex0)
                   and set(_lex[len(_lex0):]) <= _MARKERS)
            _r["bhsa_shaped"] += _ok
            _r["pos_marker"] += bool(_lex) and (_lex.endswith("[") or "/" in _lex)
            _r["homonym_marker"] += "=" in (_lex or "")
            # A word that carries a lexeme but reaches no lexeme node would be one whose lexical
            # identity nothing published can state. This is the assertion that makes withdrawing
            # the lexeme identifier payable.
            _r["lexeme_without_node"] += _lx is None

table("TEST 12  coverage of the lexical features", test12,
      ["words", "lexeme", "no_lexeme", "lex_utf8", "lexeme_node"],
      pct=[("% with a lexeme", "lexeme", "words"),
           ("% reaching a lexeme node", "lexeme_node", "words")])
print()
table("TEST 12  the disambiguation shape", test12,
      ["lexeme", "bhsa_shaped", "pos_marker", "homonym_marker", "lexeme_without_node"],
      pct=[("% carrying a marker", "pos_marker", "lexeme")])
print()

_t12 = Counter()
for _c in CORPORA:
    _t12.update(test12[_c])
check("`lex` not built on `lex0`", _t12["lexeme"] - _t12["bhsa_shaped"], 0,
      "every lexeme must be `lex0` followed only by characters drawn from `[`, `/` and `=`")
check("lexeme-bearing words that reach no `lexeme` node", _t12["lexeme_without_node"], 0,
      "containment is the only edge from a word to its lexeme, so a word that reaches no "
      "lexeme node has no published lexical identity")

# `rank_lex` carries the identity of the lexeme tier, which is why the tier needs no separate
# key. The ranking is total rather than dense, so it is a bijection onto the lexeme nodes, and
# the assertion is that property rather than mere presence: a dense ranking would give every
# hapax legomenon the same value and would identify nothing.
_ranks = [F.rank_lex.v(_lx) for _lx in F.otype.s("lexeme")]
_n_lexemes = len(_ranks)
print(f"  {_n_lexemes:,} lexeme nodes, ranks {min(_ranks):,} to {max(_ranks):,}, "
      f"{len(set(_ranks)):,} distinct")
check("lexeme nodes sharing a `rank_lex`", _n_lexemes - len(set(_ranks)), 0,
      "a total ranking is a bijection onto the lexeme nodes, which is what allows the tier to "
      "be keyed on it")
check("gaps in the `rank_lex` sequence", sorted(_ranks) != list(range(1, _n_lexemes + 1)), False,
      "the ranking must cover 1 to the number of lexemes without a gap")

# `freq_lex` is counted over the four distinct witnesses and deliberately not over the base view
# of Neofiti, which is a second view of a manuscript already counted. A lexeme attested only in
# that view therefore has a frequency of 0, which is the honest value rather than a missing one.
_freqs = [F.freq_lex.v(_lx) for _lx in F.otype.s("lexeme")]
print(f"  {sum(_freqs):,} tokens counted across {_n_lexemes:,} lexemes, "
      f"{sum(1 for _f in _freqs if _f == 0)} of them attested only in the base view")
check("lexeme nodes with no `freq_lex`", sum(1 for _f in _freqs if _f is None), 0,
      "the frequency is counted for every lexeme, and 0 is a value rather than an absence")
print()

# The manuscript tier. `text` names the manuscript everywhere except Neofiti, where two views
# describe one codex, so the collapse from 46 text nodes to 45 manuscripts is the whole content
# of this feature and is what test 14 relies on to tell a parallel from a view.
_ms = Counter(F.manuscript.v(_t) for _t in F.otype.s("text"))
_n_texts = sum(_ms.values())
print(f"  {_n_texts} text nodes over {len(_ms)} manuscripts")
for _m, _n in sorted(_ms.items(), key=lambda kv: (-kv[1], kv[0] or "")):
    if _n > 1:
        print(f"    {str(_m):<34} {_n:>4} text nodes")
check("text nodes with no `manuscript`", _ms[None], 0,
      "every text node names the manuscript it witnesses")
check("manuscripts carrying more than one text node",
      sum(1 for _n in _ms.values() if _n > 1), 1,
      "exactly one manuscript is published under two views, and that is Neofiti 1")
print()

results["12 the features of this version say what they claim"] = (
    _t12["words"], _t12["words"] - (_t12["lexeme"] - _t12["bhsa_shaped"])
                                 - _t12["lexeme_without_node"])
```

```python
# TEST 13. The lexeme against the surface, which is where `lex` and `cons` disagree about letters.
#
# The method is to ask, for every word that has both a lexeme and a surface form, whether any
# letter of `lex0` appears in `cons`. Two populations emerge, and they are unlike each other in
# both size and kind, which is why they are reported separately rather than as one failure count.
#
# The first is a letter the corpus had lost. The lemma identifiers of the source edition are
# lowercase, and its transliteration distinguishes tet from taw only by case, so a lowercased
# identifier cannot spell tet without colliding and the source substitutes another character for
# it. The character map used when the corpus was built held no entry for that substitute, so the
# conversion skipped it silently and the stored headword came out one letter short. An Aramaic
# corpus in which no lexeme and no root holds a tet is not a corpus with a rare letter. The fault
# has since been corrected in the transliteration and backfilled across the stored lemmas, so
# this cell is now a regression guard rather than the original measurement.
#
# The assertion that the guard rests on has had to change, because it previously compared the
# published letter against the lemma identifier of the source, and that identifier is no longer
# part of the dataset. Two measurements replace it, and between them they cover more than the
# original did.
#
# Within the dataset, the identity becomes one that holds across two tiers. `lex0` is published
# on the lexeme tier as well as the word tier, so the question of whether the lemma spells a tet
# where the lexeme of the token does is answerable from published values alone. Both counts stand
# at 12,789, meaning the two tiers agree about the letter on every word that carries it, which is
# a stronger statement than the original, because it holds over values a consumer can read rather
# than over a key a consumer cannot resolve.
#
# Within the database, the classification that needs the identifier of the source remains in
# place and is still run, as four checks in the SQL baseline. They establish the reconstruction
# from the source lexicon with no rows left over, count what the missing letter cost, decompose
# the lemma that an analysed unit did not earn, and list the identifiers that resolve to nothing.
# That is the correct side of the boundary for them, since a question about an internal
# identifier belongs to the check that runs against the database rather than to the notebook that
# measures the published artefact.
#
# The second population is a property of the lexical model and must not be read as a defect at
# all. The source attaches the headword of an entry to every analysed unit the entry covers, so a
# pronominal suffix or a one-letter clitic carries the lexeme of the word it attaches to. Of the
# words that share no letter with their lexeme, the decomposition below assigns each to exactly
# one cause, in order, and only the final row is unexplained. Most of that row is a one-letter
# surface.
#
# The predicates of the decomposition are drawn from published features only, which has cost it
# two rows. A compound lexical entry has no published proxy at all. A multi-word entry no longer
# has one either, now that the ordinal of the segmentation model is withheld: the only candidate
# predicate available, a word sharing its group, lexeme and part with another, admits 6,941 false
# positives against 5,636 true ones, so no proxy is offered in place of the measurement. Those
# words now fall through to the sibling, neighbour and unexplained rows, which is why the
# unexplained row is reported rather than pinned. A lemma spelled with a tet survives as a row,
# because the letter is now carried by `lex0` itself.

_TET = "V"                       # The ETCBC tet in this transliteration scheme.
_NOT_ETCBC = re.compile(r"[^A-Z><]")


def etcbc_letters(s):
    """The comparison class for this test: ETCBC consonants only.

    `cons` can also hold the lacuna dots, and that is not a letter the lexeme could be expected
    to share.
    """
    return _NOT_ETCBC.sub("", s or "")


test13 = defaultdict(Counter)
_cand = {}                       # word -> (corpus, lex0 letters, cons letters)
for _k, _vns in verse_nodes.items():
    _r = test13[corpus_of_verse[_k]]
    for _v in _vns:
        for _w in L.d(_v, otype="word"):
            _lex0, _cons = F.lex0.v(_w), etcbc_letters(F.cons.v(_w))
            _lx = lexeme_of.get(_w)
            _r["words"] += 1
            _r["tet_in_cons"] += _TET in _cons
            _r["tet_in_lex0"] += _TET in (_lex0 or "")
            _r["tet_in_root"] += _TET in (F.root.v(_w) or "")
            _r["tet_in_lexeme_tier"] += bool(_lx and _TET in (F.lex0.v(_lx) or ""))
            if not _lex0 or not _cons:
                continue
            _r["comparable"] += 1
            if set(_lex0) & set(_cons):
                continue
            _r["shares_no_letter"] += 1
            _cand[_w] = (corpus_of_verse[_k], _lex0, _cons)

table("TEST 13  the tet the source substitutes, and what the lexeme kept of it", test13,
      ["words", "tet_in_cons", "tet_in_lex0", "tet_in_lexeme_tier", "tet_in_root"],
      pct=[("% of words whose lexeme spells one", "tet_in_lex0", "words")])
print()

# The decomposition. Each word falls in the first bucket that claims it, so the rows sum to
# `shares_no_letter` and no word is counted twice.
#
# These rows do not match `baseline.sql` 16c's, and the reason is the node model rather than a
# disagreement: a Text-Fabric `word_group` node is re-opened whenever its parent changes, so one
# published `word_group_id` can be several nodes here, and a sibling that SQL finds inside the
# group is found by the segment test instead. The total, and the transposition count the campaign
# rests on, are identical either way.
_BUCKETS = ["an abbreviation", "a lemma spelled with a tet",
            "the lemma of a sibling word", "the lemma of a neighbour", "unexplained"]
_why = defaultdict(Counter)
_transposed = set()
for _w, (_c, _lex0, _cons) in _cand.items():
    _group = (L.u(_w, otype="word_group") or (None,))[0]
    _siblings = [s for s in (L.d(_group, otype="word") if _group else []) if s != _w]
    _in_sibling = any(set(_lex0) <= set(etcbc_letters(F.cons.v(s))) for s in _siblings)
    # A transposition, meaning that the lexeme of this word is found in the surface of a sibling
    # while the lexeme of that sibling is found in this one, and neither is found in its own. No
    # process in the lexical model produces a mutual exchange of that kind.
    for _s in _siblings:
        if _s in _cand and set(_lex0) <= set(etcbc_letters(F.cons.v(_s))) \
                and set(F.lex0.v(_s) or "") <= set(_cons):
            _transposed.add(_w)
    if "abbreviation" in (F.gloss.v(_w) or ""):
        _b = _BUCKETS[0]
    elif _TET in _lex0:
        _b = _BUCKETS[1]
    elif _in_sibling:
        _b = _BUCKETS[2]
    else:
        _segment = (L.u(_w, otype="segment") or (None,))[0]
        _near = [n for n in (L.d(_segment, otype="word") if _segment else [])
                 if (L.u(n, otype="word_group") or (None,))[0] != _group]
        _b = _BUCKETS[3] if any(set(_lex0) <= set(etcbc_letters(F.cons.v(n))) for n in _near) \
             else _BUCKETS[4]
    _why[_b][_c] += 1
    _why[_b]["words"] += 1

_t13 = Counter()
for _c in CORPORA:
    _t13.update(test13[_c])
print("TEST 13  the words that share no letter with their own lexeme, by cause")
print(f"  {'cause':<32} {'words':>8}")
print("  " + "-" * 72)
for _b in _BUCKETS:
    print(f"  {_b:<32} {_why[_b]['words']:>8,}")
print("  " + "-" * 72)
print(f"  {'TOTAL':<32} {sum(_why[_b]['words'] for _b in _BUCKETS):>8,}"
      f"   against shares_no_letter {_t13['shares_no_letter']:,}")
print()
_one_letter = sum(1 for _c, _l, _s in _cand.values() if len(_s) == 1)
print(f"  {_one_letter:,} of the {_t13['shares_no_letter']:,} have a one-letter surface, "
      f"which is what a clitic or a pronominal suffix is")
_groups = {(L.u(_w, otype='word_group') or (None,))[0] for _w in _transposed}
print(f"  {len(_transposed):,} words in {len(_groups):,} word groups have their lemmas transposed")
print()

check("words whose lexeme holds a tet", _t13["tet_in_lex0"], 12789,
      "0 before the correction. Measured on the word tier against the 496,532-slot build")
check("the lexeme tier disagreeing with the word tier about the letter",
      _t13["tet_in_lex0"] - _t13["tet_in_lexeme_tier"], 0,
      "the identity shared across the two tiers. `lex0` is published on `word` and on `lexeme`, "
      "and the two must spell the same lemma")
check("words whose root holds a tet", _t13["tet_in_root"], 12199,
      "0 before the correction. Lower than `lex0` because the root covers fewer units, not "
      "because a letter was lost")
check("words sharing no letter with their lexeme", _t13["shares_no_letter"], 8487,
      "a documented ceiling rather than a defect count. The source attaches the headword of an "
      "unit the entry covers. Re-pinned against the 496,532-slot build")
check("word groups whose monads hold each other's lemma", len(_groups), 0,
      "all 14 were corrected in the source. This is an invariant rather than a ceiling, because "
      "no process in the lexical model produces a mutual exchange, so a new one is a defect")

# The summary row: what is left to measure is how much of the lexeme-against-surface relation the
# five rules above can ACCOUNT for, so the shortfall is the unexplained bucket and nothing else.
results["13 the lexeme belongs to its surface"] = (
    _t13["comparable"], _t13["comparable"] - _why[_BUCKETS[-1]]["words"])
```

```python
# TEST 14. The parallel witness edge, and the five properties that make it a relation rather
# than a list of pairs.
#
# The relation was first published as a set of database views and a keyed table, and an edge
# feature had to wait for a rebuild, because an edge keys on node numbers and node numbers do not
# exist until the corpus has been compiled. That rebuild has happened, so the relation is now a
# feature of the core data:
#
#     E.crossref.f(verse)  ->  ((other_verse, 'witness' | 'view' | 'same_witness'), ...)
#
# The application configuration registers no additional modules as a consequence. An earlier plan
# would have shipped this as a separate module with its own directory and identifier, and
# registering one now would direct a consumer to a second artefact for a feature they already
# hold.
#
# The method is to walk every verse node, read its edges, and assert five properties. The keyed
# table remains the citable statement of the relation, because it is keyed on the address and
# therefore survives a recompile, and a verification mode compares the two. This cell asserts the
# edge against the corpus it ships in, which is the question a consumer of the edge actually has.
#
#   1  Symmetry. The edge is emitted in both directions, so every edge has its reverse carrying
#      the same label. Without that property the edge answers from one side only and a consumer
#      has to know which side they are on.
#
#   2  The address, in both directions. Two verses are parallel when they carry the same book,
#      chapter and verse, so no edge may join two addresses and no two nodes at one address may
#      be left unjoined. The second half is what makes the edge the address relation itself
#      rather than a subset of it.
#
#   3  The label, which has three values rather than two.
#        `witness`       the manuscripts differ, and this is the parallel.
#        `view`          one manuscript under two views, which is Neofiti 1 and nothing else.
#        `same_witness`  one manuscript under one view, twice at one address.
#      This is what the manuscript feature is published for. Collapsing `witness` and `view`
#      makes the number of witnesses attesting a verse wrong by one on all 5,827 Neofiti
#      addresses. The third label is the price of the second property being total, since three
#      Genizah addresses hold more than one verse record of one fragment, which accounts for 20
#      ordered pairs. Naming that case was preferred to labelling it `view`, which would be
#      false, and to omitting it, which would make the edge under-report a consumer's own witness
#      with nothing to say so. The keyed table does not face the question, because it is keyed on
#      the manuscript and view and so excludes the pair rather than naming it.
#
#   4  NO SELF-EDGE, and both endpoints on the `verse` tier.
#
#   5  NO OTHER LABEL. Those three exhaust the relation; a fourth value would be a fourth kind
#      of thing that nothing documents.
#
# What it measures: whether a consumer can trust the edge INSTEAD of re-deriving the relation
# from the address. If any check here fails they cannot, and the CSV is the fallback.

_addr, _ms = {}, {}
for _v in F.otype.s("verse"):
    _t = L.u(_v, otype="text")[0]
    _addr[_v] = (F.book.v(L.u(_v, otype="book")[0]),
                 F.chapter.v(L.u(_v, otype="chapter")[0]),
                 _verse_number(_v))
    _ms[_v] = (F.manuscript.v(_t), F.view.v(_t))

_at = defaultdict(list)
for _v, _a in _addr.items():
    _at[_a].append(_v)

RELATIONS = ("witness", "view", "same_witness")
_edges, _rel = {}, Counter()
for _v in F.otype.s("verse"):
    for _o, _r in E.crossref.f(_v) or ():
        _edges[(_v, _o)] = _r
        _rel[_r] += 1


def _faults(a, b, r):
    """Which of the five properties this one ordered edge breaks."""
    out = []
    if _edges.get((b, a)) != r:
        out.append("asymmetric")
    if _addr.get(b) != _addr[a]:
        out.append("off-address")
    if r != ("witness" if _ms[b][0] != _ms[a][0]
             else "view" if _ms[b][1] != _ms[a][1] else "same_witness"):
        out.append("mislabelled")
    if a == b:
        out.append("self")
    if F.otype.v(b) != "verse":
        out.append("off-tier")
    if r not in RELATIONS:
        out.append("unknown-relation")
    return out


_broken = Counter()
_good = 0
for (_a, _b), _r in _edges.items():
    _f = _faults(_a, _b, _r)
    if _f:
        _broken.update(_f)
    else:
        _good += 1

# The other direction of property 2: what the address relation REQUIRES.
_required = sum(len(_n) * (len(_n) - 1) for _n in _at.values())
_isolated = [_a for _a, _n in _at.items() if len(_n) == 1]
_shared = len(_at) - len(_isolated)
_by_witness = Counter(len({_ms[_v][0] for _v in _n}) for _n in _at.values())

print("TEST 14  the parallel-witness edge")
print(f"  {'ordered pairs in the edge':<48} {len(_edges):>9,}")
print(f"  {'  witness (two manuscripts, the parallel)':<48} {_rel['witness']:>9,}")
print(f"  {'  view (one manuscript, two view types)':<48} {_rel['view']:>9,}")
print(f"  {'  same_witness (one manuscript, one view)':<48} {_rel['same_witness']:>9,}")
print(f"  {'ordered pairs the address relation requires':<48} {_required:>9,}")
print(f"  {'canonical addresses':<48} {len(_at):>9,}")
print(f"  {'  with more than one verse node':<48} {_shared:>9,}")
print(f"  {'  attested once, so carrying no edge':<48} {len(_isolated):>9,}")
print(f"  {'verse nodes':<48} {len(_addr):>9,}")
print()
print("  addresses by the number of MANUSCRIPTS that attest them")
for _n in sorted(_by_witness, reverse=True):
    print(f"    {_n:>2} manuscript(s)  {_by_witness[_n]:>6,}")
print()

check("crossref edges breaking any of the five properties", sum(_broken.values()), 0,
      "by fault: " + (", ".join(f"{_k}={_v:,}" for _k, _v in sorted(_broken.items()))
                      or "none"))
check("ordered pairs the address relation requires but the edge omits",
      _required - len(_edges), 0,
      "the edge must BE the address relation, not a subset of it: every pair of verse nodes "
      "sharing `(book, chapter, verse)` is joined, in both directions")
check("crossref relation values outside the documented three",
      len(set(_rel) - set(RELATIONS)), 0,
      "found: " + (", ".join(sorted(set(_rel) - set(RELATIONS))) or "none"))
check("`same_witness` ordered pairs", _rel["same_witness"], 20,
      "three Geniza addresses hold more than one verse record of one fragment: "
      "`_dd` at Deuteronomy 24:6 (2 nodes), `_ff` at Exodus 15:12 (4) and 17:12 (3). A ceiling, "
      "like the sentinel population: it may legitimately be explained, never grow silently")
check("addresses attested by exactly one verse node", len(_isolated), 22,
      "a verse no other manuscript transmits and Neofiti does not hold twice, so it has no "
      "parallel to point at. Reported as a ceiling, since it may shrink as further witnesses "
      "are added")

# Four ways to use the edge, each answering a question the corpus could not answer before it.
print("  worked examples")
_ex = T.nodeFromSection(("Genesis", 44, 18))
if _ex is not None:
    _them = E.crossref.f(_ex) or ()
    print(f"    1  witnesses to Genesis 44:18 from {_ms[_ex][0]}: "
          f"{len({_ms[_o][0] for _o, _r in _them if _r == 'witness'})} other manuscript(s)")
    for _o, _r in sorted(_them, key=lambda p: _ms[p[0]]):
        print(f"         {_r:<8} {_ms[_o][0]:<22} {len(L.d(_o, otype='word')):>4} words")

# 2. The most widely attested addresses, where a critical apparatus has the most to compare.
_deep = sorted(_at.items(), key=lambda kv: -len({_ms[_v][0] for _v in kv[1]}))[:3]
print(f"    2  the three most widely attested addresses")
for _a, _n in _deep:
    print(f"         {_a[0]} {_a[1]}:{_a[2]:<4} "
          f"{len({_ms[_v][0] for _v in _n})} manuscripts, {len(_n)} verse nodes")

# 3. The `view` edges on their own: one codex read two ways, which is a different question.
_views = {_a for (_a, _b), _r in _edges.items() if _r == "view"}
print(f"    3  verse nodes with a `view` edge (Neofiti base vs full): {len(_views):,}")

# 4. Length disagreement between witnesses, the cheapest signal that a reading diverges.
_spread = []
for _a, _n in _at.items():
    if len({_ms[_v][0] for _v in _n}) < 3:
        continue
    _w = [len(L.d(_v, otype="word")) for _v in _n]
    _spread.append((max(_w) - min(_w), _a, min(_w), max(_w), len(_n)))
_spread.sort(reverse=True)
print(f"    4  widest word-count spread among 3+ manuscript addresses")
for _d, _a, _lo, _hi, _k in _spread[:3]:
    print(f"         {_a[0]} {_a[1]}:{_a[2]:<4} {_lo:>4} to {_hi:<4} words "
          f"across {_k} verse nodes (spread {_d})")

results["14 the crossref edge is the address relation"] = (_required, _good)
```
